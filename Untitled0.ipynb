```python
%cd /content/drive/MyDrive/ULTIMATEAREACALCULATOR/ComfyUI/ComfyUI

!npm install -g localtunnel

import subprocess
import threading
import time
import socket
import urllib.request

def iframe_thread(port):
    while True:
        time.sleep(0.5)
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        result = sock.connect_ex(("127.0.0.1", port))
        sock.close()

        if result == 0:
            break

    print("\nComfyUI finished loading. Launching Localtunnel...")

    process = subprocess.Popen(
        ["lt", "--port", str(port)],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    for line in process.stdout:
        print("\nCLICK THIS LINK TO OPEN COMFYUI:", line, end="")

threading.Thread(
    target=iframe_thread,
    daemon=True,
    args=(8188,)
).start()

endpoint_ip = urllib.request.urlopen(
    "https://ipv4.icanhazip.com"
).read().decode("utf8").strip()

print(f"\n---> YOUR ENDPOINT IP IS: {endpoint_ip} <---")

!python main.py --dont-print-server --enable-manager --enable-manager-legacy-ui
```
