```python
# ============================================================
# 🚀 ONE-CELL COMFYUI STARTUP
#
# DagsHub persistent storage
#        ↓
# Fast Colab storage (/content/ComfyUI)
#        ↓
# Dependency check
#        ↓
# ComfyUI on T4
#        ↓
# LocalTunnel
#
# Run ONLY this cell for normal startup.
# ============================================================

import os
import sys
import time
import socket
import shutil
import subprocess
import urllib.request
from pathlib import Path

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

REPO = "dpbasvs-afk/comfyui-project"

DAGSHUB_ROOT = Path("/content/dagshub_storage")
DAGSHUB_COMFY = DAGSHUB_ROOT / "comfyui_project" / "ComfyUI"

FAST_COMFY = Path("/content/ComfyUI")

PORT = 8188

print("=" * 65)
print("🚀 COMFYUI ONE-CELL STARTUP")
print("=" * 65)


# ------------------------------------------------------------
# 1. CHECK / RECONNECT DAGSHUB
# ------------------------------------------------------------

print("\n[1/7] Checking DagsHub storage...")

try:
    # Test whether the existing mount is actually usable.
    test_file = DAGSHUB_ROOT / "comfyui_project"

    if not DAGSHUB_ROOT.exists():
        print("DagsHub mount not found. Mounting...")

        import dagshub

        DAGSHUB_ROOT.mkdir(parents=True, exist_ok=True)

        dagshub.storage.mount(
            REPO,
            path=str(DAGSHUB_ROOT)
        )

    else:
        # The folder can exist while the FUSE mount is broken.
        try:
            list(DAGSHUB_ROOT.iterdir())
            print("✅ Existing DagsHub mount appears usable.")
        except OSError:
            print("⚠️ Existing DagsHub mount is broken.")
            print("Attempting to reconnect...")

            import dagshub

            try:
                dagshub.storage.unmount(
                    repo=REPO,
                    path=str(DAGSHUB_ROOT)
                )
            except Exception:
                pass

            time.sleep(2)

            dagshub.storage.mount(
                REPO,
                path=str(DAGSHUB_ROOT)
            )

    # Final check
    if not DAGSHUB_COMFY.exists():
        raise RuntimeError(
            f"DagsHub ComfyUI folder not found:\n{DAGSHUB_COMFY}"
        )

    print("✅ DagsHub connected")
    print("   ", DAGSHUB_COMFY)

except Exception as e:
    print("\n❌ DagsHub connection failed:")
    print(type(e).__name__, e)
    raise


# ------------------------------------------------------------
# 2. CHECK SOURCE FILES
# ------------------------------------------------------------

print("\n[2/7] Checking persistent ComfyUI...")

required_source = [
    "main.py",
    "requirements.txt",
    "comfy",
    "comfy_api",
]

for item in required_source:
    p = DAGSHUB_COMFY / item
    print(("✅" if p.exists() else "❌"), item)

if not (DAGSHUB_COMFY / "main.py").exists():
    raise RuntimeError("❌ DagsHub ComfyUI is incomplete: main.py missing.")

if not (DAGSHUB_COMFY / "comfy_api").exists():
    raise RuntimeError("❌ DagsHub ComfyUI is incomplete: comfy_api missing.")


# ------------------------------------------------------------
# 3. SYNC TO FAST COLAB STORAGE
# ------------------------------------------------------------

print("\n[3/7] Syncing DagsHub → fast Colab storage...")
print("This is the important speed step.")

FAST_COMFY.mkdir(parents=True, exist_ok=True)

result = subprocess.run(
    [
        "rsync",
        "-a",
        "--exclude=.git/",
        str(DAGSHUB_COMFY) + "/",
        str(FAST_COMFY) + "/",
    ],
    text=True,
    capture_output=True
)

if result.returncode != 0:
    print(result.stdout[-3000:])
    print(result.stderr[-3000:])
    raise RuntimeError("❌ Fast-storage sync failed.")

print("✅ Sync completed")


# ------------------------------------------------------------
# 4. VERIFY FAST COPY
# ------------------------------------------------------------

print("\n[4/7] Verifying fast ComfyUI...")

required_fast = [
    "main.py",
    "requirements.txt",
    "comfy",
    "comfy/sd.py",
    "comfy/ldm/models",
    "comfy_api",
    "comfy_api/input",
    "comfy_api/input/__init__.py",
    "comfy_api/latest",
    "comfy_api/latest/_io.py",
]

missing = []

for item in required_fast:
    p = FAST_COMFY / item

    if p.exists():
        print("✅", item)
    else:
        print("❌", item)
        missing.append(item)

if missing:
    raise RuntimeError(
        "\n❌ FAST COPY IS INCOMPLETE:\n" +
        "\n".join(" - " + x for x in missing)
    )

print("✅ Fast copy is complete")


# ------------------------------------------------------------
# 5. PYTHON PATH + DEPENDENCIES
# ------------------------------------------------------------

print("\n[5/7] Preparing Python environment...")

os.chdir(FAST_COMFY)

if str(FAST_COMFY) not in sys.path:
    sys.path.insert(0, str(FAST_COMFY))

# Install ComfyUI requirements.
# pip will normally skip packages that are already satisfied.
print("Checking ComfyUI dependencies...")

pip_result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-r",
        str(FAST_COMFY / "requirements.txt"),
        "-q",
    ],
    text=True,
    capture_output=True
)

if pip_result.returncode != 0:
    print("⚠️ Dependency installation reported an error:")
    print(pip_result.stderr[-5000:])
    raise RuntimeError("Dependency installation failed.")

print("✅ Dependencies ready")


# ------------------------------------------------------------
# 6. START COMFYUI
# ------------------------------------------------------------

print("\n[6/7] Starting ComfyUI...")

def port_open(port):
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(1)
    try:
        return s.connect_ex(("127.0.0.1", port)) == 0
    finally:
        s.close()


# Check whether an old ComfyUI process is already using 8188.
if port_open(PORT):
    print(f"⚠️ Port {PORT} is already open.")
    print("Assuming ComfyUI is already running.")

    comfy_process = None

else:
    log_path = FAST_COMFY / "comfyui_startup.log"

    log_file = open(log_path, "w")

    env = os.environ.copy()
    env["PYTHONUNBUFFERED"] = "1"

    comfy_process = subprocess.Popen(
        [
            sys.executable,
            "-u",
            "main.py",
            "--listen",
            "0.0.0.0",
            "--port",
            str(PORT),
            "--cuda-device",
            "0",
        ],
        cwd=str(FAST_COMFY),
        stdout=log_file,
        stderr=subprocess.STDOUT,
        env=env,
        start_new_session=True,
    )

    print("ComfyUI PID:", comfy_process.pid)
    print("Waiting for ComfyUI to finish loading...")

    # Up to 20 minutes for a cold start.
    deadline = time.time() + 1200

    while time.time() < deadline:

        if port_open(PORT):
            print("\n🎉 COMFYUI SERVER IS READY!")
            break

        if comfy_process.poll() is not None:
            log_file.close()

            print("\n❌ ComfyUI exited before becoming ready.")
            print("\n===== LAST STARTUP LOG =====")

            try:
                print(log_path.read_text()[-10000:])
            except Exception:
                pass

            raise RuntimeError("ComfyUI failed to start.")

        time.sleep(2)

    else:
        log_file.close()
        raise TimeoutError(
            "ComfyUI did not open port 8188 within 20 minutes."
        )


# ------------------------------------------------------------
# 7. LOCAL TUNNEL
# ------------------------------------------------------------

print("\n[7/7] Starting LocalTunnel...")

# Install Node/localtunnel only if needed.
lt_check = subprocess.run(
    ["bash", "-lc", "command -v lt"],
    capture_output=True,
    text=True
)

if lt_check.returncode != 0:
    print("Installing localtunnel...")
    install_lt = subprocess.run(
        ["npm", "install", "-g", "localtunnel"],
        text=True,
        capture_output=True
    )

    if install_lt.returncode != 0:
        print(install_lt.stderr[-5000:])
        raise RuntimeError("LocalTunnel installation failed.")

print("✅ LocalTunnel installed")


# Get the endpoint IP used by LocalTunnel's password page.
try:
    endpoint_ip = urllib.request.urlopen(
        "https://ipv4.icanhazip.com",
        timeout=10
    ).read().decode().strip()

    print("\nLocalTunnel endpoint/password IP:")
    print(endpoint_ip)

except Exception as e:
    endpoint_ip = None
    print("⚠️ Could not retrieve LocalTunnel endpoint IP:", e)


# Start tunnel
tunnel_process = subprocess.Popen(
    ["lt", "--port", str(PORT)],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

print("\n⏳ Waiting for LocalTunnel URL...")

tunnel_url = None
tunnel_deadline = time.time() + 60

while time.time() < tunnel_deadline:

    line = tunnel_process.stdout.readline()

    if line:
        line = line.strip()

        if line:
            print(line)

        if "https://" in line:
            start = line.find("https://")
            tunnel_url = line[start:].split()[0]
            break

    elif tunnel_process.poll() is not None:
        break

    else:
        time.sleep(0.2)


# ------------------------------------------------------------
# FINAL RESULT
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("🎉 COMFYUI STARTUP COMPLETE")
print("=" * 65)

print("\nFast ComfyUI:")
print(FAST_COMFY)

print("\nComfyUI local server:")
print(f"http://127.0.0.1:{PORT}")

if tunnel_url:
    print("\n🌐 COMFYUI PUBLIC URL:")
    print(tunnel_url)
else:
    print("\n⚠️ LocalTunnel did not provide a URL automatically.")
    print("Check the LocalTunnel output above.")

if endpoint_ip:
    print("\nIf LocalTunnel asks for its endpoint/password IP, use:")
    print(endpoint_ip)

print("\n🚀 Keep this cell running while using ComfyUI.")
print("=" * 65)
```
