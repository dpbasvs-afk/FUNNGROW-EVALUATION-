STARTUP CELL

```python
# ==========================================
# COMFYUI + DAGSHUB STARTUP
# ==========================================

# 1. Install DagsHub
%pip install -q -U "dagshub[jupyter]"

# 2. Authenticate
import dagshub

dagshub.auth.add_oauth_token()

# 3. Connect to your repository
from pathlib import Path

REPO = "dpbasvs-afk/comfyui-project"
MOUNT_PATH = Path("/content/dagshub_storage")

mount_path = dagshub.storage.mount(
    REPO,
    path=MOUNT_PATH,
    cache=True
)

# 4. Recreate persistent paths
PERSISTENT_ROOT = Path(mount_path) / "comfyui_project"
PERSISTENT_ROOT.mkdir(parents=True, exist_ok=True)

COMFY_DIR = PERSISTENT_ROOT / "ComfyUI"

print("================================")
print("DagsHub connected!")
print("Persistent root:", PERSISTENT_ROOT)
print("ComfyUI path:", COMFY_DIR)
print("ComfyUI exists:", (COMFY_DIR / "main.py").exists())
print("================================")
```

```python
import subprocess
import sys

requirements_file = COMFY_DIR / "requirements.txt"

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-r",
        str(requirements_file)
    ],
    capture_output=False
)

print("Dependency installation finished with code:", result.returncode)
```

INSTALL DEPENDICIES


```python
import subprocess
import sys

result = subprocess.run(
    [
        sys.executable,
        str(COMFY_DIR / "main.py"),
        "--help"
    ],
    cwd=str(COMFY_DIR),
    capture_output=True,
    text=True,
    timeout=120
)

print("Return code:", result.returncode)
print(result.stdout[-3000:])

if result.stderr:
    print("\nErrors / warnings:")
    print(result.stderr[-2000:])
```

```python
import shutil
from pathlib import Path

FAST_COMFY = Path("/content/ComfyUI")

if not FAST_COMFY.exists():
    print("Copying ComfyUI to fast Colab storage...")
    shutil.copytree(
        COMFY_DIR,
        FAST_COMFY,
        ignore=shutil.ignore_patterns(
            ".git",
            "__pycache__",
            "output",
            "temp",
            "models"
        )
    )
    print("Copy completed!")
else:
    print("Fast ComfyUI copy already exists.")

print("ComfyUI location:", FAST_COMFY)
print("main.py exists:", (FAST_COMFY / "main.py").exists())
```
