```python
# ================================================================
# 🚀 JULIUS CAESAR — PERMANENT EXTENSION SETUP
# ================================================================
# RUN ORDER:
#   1. MASTER STARTUP CELL
#   2. THIS CELL  ← extensions/dependencies
#   3. COMFYUI + MANAGER START CELL (if your startup doesn't do it)
#   4. TUNNEL CELL
#
# This cell is SAFE TO RUN AGAIN.
# Existing repositories are NOT recloned.
# ================================================================

import os
import sys
import subprocess
from pathlib import Path

print("=" * 70)
print("🎬 JULIUS CAESAR — EXTENSION SETUP")
print("=" * 70)

# ------------------------------------------------
# 1. LOCATIONS
# ------------------------------------------------

COMFY = Path("/content/ComfyUI")
CUSTOM = COMFY / "custom_nodes"

if not COMFY.exists():
    raise RuntimeError(f"❌ ComfyUI not found: {COMFY}")

CUSTOM.mkdir(parents=True, exist_ok=True)

print(f"\nComfyUI : {COMFY}")
print(f"Custom nodes : {CUSTOM}")

# ------------------------------------------------
# 2. REPOSITORIES
# ------------------------------------------------
# Official repositories for the production stack.
#
# We are installing:
#   AnimateDiff Evolved
#   VideoHelperSuite
#   IPAdapter Plus
#   Advanced ControlNet
#   ControlNet Aux
#   Impact Pack
#   TeaCache
#
# F5-TTS and LivePortrait are handled separately because
# they are not simply ordinary ComfyUI custom-node folders.
# ------------------------------------------------

REPOS = {
    "ComfyUI-AnimateDiff-Evolved":
        "https://github.com/Kosinkadink/ComfyUI-AnimateDiff-Evolved.git",

    "ComfyUI-VideoHelperSuite":
        "https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite.git",

    "ComfyUI_IPAdapter_plus":
        "https://github.com/cubiq/ComfyUI_IPAdapter_plus.git",

    "ComfyUI-Advanced-ControlNet":
        "https://github.com/Kosinkadink/ComfyUI-Advanced-ControlNet.git",

    "comfyui_controlnet_aux":
        "https://github.com/Fannovel16/comfyui_controlnet_aux.git",

    "ComfyUI-Impact-Pack":
        "https://github.com/ltdrdata/ComfyUI-Impact-Pack.git",

    "ComfyUI-TeaCache":
        "https://github.com/nvmax/teacache.git",
}

# ------------------------------------------------
# 3. GIT INSTALLER
# ------------------------------------------------

def run_cmd(cmd, cwd=None):
    print("\n$", " ".join(map(str, cmd)))
    result = subprocess.run(
        cmd,
        cwd=str(cwd) if cwd else None,
        text=True
    )
    if result.returncode != 0:
        print(f"⚠️ Command returned {result.returncode}")
    return result.returncode


def install_or_update(name, url):

    target = CUSTOM / name

    print("\n" + "-" * 65)
    print(f"📦 {name}")
    print("-" * 65)

    if target.exists() and (target / ".git").exists():

        print("✅ Already cloned")

        # Do NOT blindly pull every time.
        # The current checkout is preserved to avoid unexpected
        # changes breaking a working production environment.

        print("ℹ️ Existing checkout preserved.")

    elif target.exists():

        print("⚠️ Folder exists but is not a Git repository.")
        print("Keeping it untouched:", target)

    else:

        print("⬇️ Cloning...")
        rc = run_cmd(
            ["git", "clone", "--depth", "1", url, str(target)]
        )

        if rc == 0:
            print("✅ Clone complete")
        else:
            print("❌ Clone failed")

    # ------------------------------------------------
    # Install extension-specific Python requirements
    # ------------------------------------------------

    requirements = target / "requirements.txt"

    if requirements.exists():

        print(f"📚 Installing requirements for {name}...")

        rc = run_cmd(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-r",
                str(requirements)
            ]
        )

        if rc == 0:
            print("✅ Requirements OK")
        else:
            print("⚠️ Some requirements may have failed.")

    else:
        print("ℹ️ No requirements.txt")

# ------------------------------------------------
# 4. INSTALL EVERYTHING
# ------------------------------------------------

for name, url in REPOS.items():
    install_or_update(name, url)

# ------------------------------------------------
# 5. MOVIEPY
# ------------------------------------------------

print("\n" + "=" * 65)
print("🎞️ MOVIEPY")
print("=" * 65)

run_cmd([
    sys.executable,
    "-m",
    "pip",
    "install",
    "moviepy"
])

print("✅ MoviePy installation attempted")

# ------------------------------------------------
# 6. VERIFY EXTENSIONS
# ------------------------------------------------

print("\n" + "=" * 70)
print("🔍 VERIFYING EXTENSIONS")
print("=" * 70)

for name in REPOS:
    path = CUSTOM / name

    if path.exists():
        print(f"✅ {name}")
    else:
        print(f"❌ {name}")

# ------------------------------------------------
# 7. VERIFY IMPORTANT COMFYUI DIRECTORIES
# ------------------------------------------------

print("\n" + "=" * 70)
print("📁 VERIFYING MODEL DIRECTORIES")
print("=" * 70)

MODEL_DIRS = [
    COMFY / "models" / "checkpoints",
    COMFY / "models" / "controlnet",
    COMFY / "models" / "clip_vision",
    COMFY / "models" / "animatediff_models",
    COMFY / "models" / "ipadapter",
    COMFY / "models" / "vae",
]

for path in MODEL_DIRS:
    path.mkdir(parents=True, exist_ok=True)
    print(f"✅ {path}")

# ------------------------------------------------
# 8. FINAL STATUS
# ------------------------------------------------

print("\n" + "=" * 70)
print("🎬 JULIUS CAESAR EXTENSION SETUP COMPLETE")
print("=" * 70)

print("""
Installed / prepared:

✅ AnimateDiff Evolved
✅ VideoHelperSuite
✅ IPAdapter Plus
✅ Advanced ControlNet
✅ ControlNet Aux
✅ Impact Pack
✅ TeaCache
✅ MoviePy

Prepared model folders:

✅ checkpoints
✅ controlnet
✅ clip_vision
✅ animatediff_models
✅ ipadapter
✅ vae

NOT downloaded automatically yet:

⬇️ Anime checkpoint
⬇️ AnimateDiff motion model
⬇️ IPAdapter model
⬇️ CLIP Vision model
⬇️ ControlNet model(s)
⬇️ F5-TTS model
⬇️ LivePortrait model

Those are MODEL WEIGHTS, not extensions.

IMPORTANT:
Restart ComfyUI after installing/updating custom nodes.
Do not start the tunnel until ComfyUI has restarted successfully.
""")

print("=" * 70)
print("✅ ONE-CELL EXTENSION SETUP FINISHED")
print("=" * 70)
```

```python
# ======================================================================
# 🎬 JULIUS CAESAR — MODEL WEIGHTS SETUP
# ======================================================================
# RUN AFTER:
#   1. MASTER STARTUP
#   2. EXTENSION SETUP
#
# This cell is SAFE TO RUN AGAIN.
# Existing files are NOT downloaded again.
#
# CURRENT STACK:
#   SD 1.5
#   AnimateDiff Evolved
#   IPAdapter Plus
#   CLIP Vision ViT-H
#
# ======================================================================

import os
import sys
import hashlib
import urllib.request
from pathlib import Path

print("=" * 70)
print("🎬 JULIUS CAESAR — MODEL WEIGHTS SETUP")
print("=" * 70)

# ----------------------------------------------------------------------
# COMFYUI
# ----------------------------------------------------------------------

COMFY = Path("/content/ComfyUI")

if not COMFY.exists():
    raise RuntimeError(f"❌ ComfyUI not found: {COMFY}")

print(f"ComfyUI: {COMFY}")

# ----------------------------------------------------------------------
# MODEL DIRECTORIES
# ----------------------------------------------------------------------

DIRS = {
    "checkpoints": COMFY / "models" / "checkpoints",
    "vae": COMFY / "models" / "vae",
    "animatediff": COMFY / "models" / "animatediff_models",
    "ipadapter": COMFY / "models" / "ipadapter",
    "clip_vision": COMFY / "models" / "clip_vision",
    "controlnet": COMFY / "models" / "controlnet",
    "loras": COMFY / "models" / "loras",
}

print("\n📁 Creating model directories...")

for name, path in DIRS.items():
    path.mkdir(parents=True, exist_ok=True)
    print(f"✅ {name}: {path}")

# ----------------------------------------------------------------------
# DOWNLOAD FUNCTION
# ----------------------------------------------------------------------

def download_if_missing(url, destination, expected_sha256=None):

    destination = Path(destination)

    if destination.exists():

        size_mb = destination.stat().st_size / (1024 * 1024)

        if size_mb > 1:
            print(f"✅ Already exists: {destination.name} ({size_mb:.1f} MB)")
            return True

        print(f"⚠️ Existing file looks incomplete: {destination.name}")
        print("Removing incomplete file...")
        destination.unlink()

    print("\n⬇️ Downloading:")
    print(destination.name)
    print("From:", url)

    try:

        urllib.request.urlretrieve(url, destination)

    except Exception as e:

        print("❌ Download failed:", e)

        if destination.exists():
            destination.unlink()

        return False

    if not destination.exists():
        print("❌ File was not created.")
        return False

    size_mb = destination.stat().st_size / (1024 * 1024)

    print(f"✅ Downloaded: {size_mb:.1f} MB")

    # --------------------------------------------------------------
    # Optional SHA256 verification
    # --------------------------------------------------------------

    if expected_sha256:

        print("🔐 Verifying SHA256...")

        sha = hashlib.sha256()

        with open(destination, "rb") as f:
            while True:
                chunk = f.read(1024 * 1024)

                if not chunk:
                    break

                sha.update(chunk)

        actual = sha.hexdigest()

        if actual.lower() != expected_sha256.lower():

            print("❌ SHA256 MISMATCH")
            print("Expected:", expected_sha256)
            print("Actual:  ", actual)

            destination.unlink()

            return False

        print("✅ SHA256 verified")

    return True


# ======================================================================
# 1. ANIMATEDIFF MOTION MODEL
# ======================================================================

print("\n" + "=" * 70)
print("🎞️ 1/4 — ANIMATEDIFF MOTION MODEL")
print("=" * 70)

ANIMATEDIFF_URL = (
    "https://huggingface.co/guoyww/animatediff/resolve/main/"
    "mm_sd_v15_v2.safetensors"
)

ANIMATEDIFF_FILE = DIRS["animatediff"] / "mm_sd_v15_v2.safetensors"

download_if_missing(
    ANIMATEDIFF_URL,
    ANIMATEDIFF_FILE,
    expected_sha256=(
        "1cf5d6d6062b8d3d72cd5cdcabb578befd79b7a3e1ea41bf15d3ca594c0998da"
    )
)

# This motion module is 1.82 GB according to its Hugging Face listing.
# It is specifically an SD1.5 AnimateDiff motion module.
# ======================================================================


# ======================================================================
# 2. IPADAPTER PLUS SD1.5
# ======================================================================

print("\n" + "=" * 70)
print("🧑 2/4 — IPADAPTER PLUS SD1.5")
print("=" * 70)

IPADAPTER_URL = (
    "https://huggingface.co/h94/IP-Adapter/resolve/main/"
    "models/ip-adapter-plus_sd15.safetensors"
)

IPADAPTER_FILE = DIRS["ipadapter"] / "ip-adapter-plus_sd15.safetensors"

download_if_missing(
    IPADAPTER_URL,
    IPADAPTER_FILE,
    expected_sha256=(
        "a1c250be40455cc61a43da1201ec3f1edaea71214865fb47f57927e06cbe4996"
    )
)

# ======================================================================
# 3. IPADAPTER PLUS FACE
# ======================================================================

print("\n" + "=" * 70)
print("🙂 3/4 — IPADAPTER PLUS FACE")
print("=" * 70)

IPADAPTER_FACE_URL = (
    "https://huggingface.co/h94/IP-Adapter/resolve/main/"
    "models/ip-adapter-plus-face_sd15.safetensors"
)

IPADAPTER_FACE_FILE = (
    DIRS["ipadapter"] /
    "ip-adapter-plus-face_sd15.safetensors"
)

download_if_missing(
    IPADAPTER_FACE_URL,
    IPADAPTER_FACE_FILE,
    expected_sha256=(
        "1c9edc21af6f737dc1d6e0e734190e976cfacf802d6b024b77aa3be922f7569b"
    )
)

# ======================================================================
# 4. CLIP VISION — ViT-H
# ======================================================================

print("\n" + "=" * 70)
print("👁️ 4/4 — CLIP VISION ViT-H")
print("=" * 70)

CLIP_VISION_URL = (
    "https://huggingface.co/h94/IP-Adapter/resolve/main/"
    "models/image_encoder/model.safetensors"
)

CLIP_VISION_FILE = (
    DIRS["clip_vision"] /
    "CLIP-ViT-H-14-laion2B-s32B-b79K.safetensors"
)

download_if_missing(
    CLIP_VISION_URL,
    CLIP_VISION_FILE
)

# ======================================================================
# CONTROLNET
# ======================================================================

print("\n" + "=" * 70)
print("🎯 CONTROLNET")
print("=" * 70)

print("""
ControlNet models are NOT being blindly downloaded yet.

Reason:
The exact ControlNet checkpoint must match the final SD1.5
anime checkpoint and the controls we actually use.

The ControlNet-Aux extension is already installed and can provide
the preprocessing nodes.

We will add only the ControlNet weights actually required by
the Caesar workflow.
""")

# ======================================================================
# BASE ANIME CHECKPOINT
# ======================================================================

print("=" * 70)
print("🎨 BASE ANIME CHECKPOINT")
print("=" * 70)

print("""
⚠️ NOT DOWNLOADED YET.

This is intentional.

We need to choose ONE exact SD1.5 anime checkpoint for the Caesar
visual style before locking the rest of the generation stack.

Do NOT put a random .safetensors file here.

Once the checkpoint is chosen, it will go into:

    /content/ComfyUI/models/checkpoints/

The rest of this cell is already prepared for it.
""")

# ======================================================================
# FINAL VERIFICATION
# ======================================================================

print("\n" + "=" * 70)
print("🔍 FINAL MODEL CHECK")
print("=" * 70)

def show_file(label, path):

    if path.exists():

        size_gb = path.stat().st_size / (1024 ** 3)

        print(f"✅ {label}")
        print(f"   {path}")
        print(f"   {size_gb:.2f} GB")

    else:

        print(f"❌ {label}")
        print(f"   Missing: {path}")


show_file(
    "AnimateDiff motion model",
    ANIMATEDIFF_FILE
)

show_file(
    "IPAdapter Plus",
    IPADAPTER_FILE
)

show_file(
    "IPAdapter Plus Face",
    IPADAPTER_FACE_FILE
)

show_file(
    "CLIP Vision ViT-H",
    CLIP_VISION_FILE
)

print("\n" + "=" * 70)
print("🎬 MODEL INFRASTRUCTURE SETUP COMPLETE")
print("=" * 70)

print("""
READY:

✅ AnimateDiff SD1.5 motion model
✅ IPAdapter Plus SD1.5
✅ IPAdapter Plus Face SD1.5
✅ CLIP Vision ViT-H
✅ Required model folders

STILL TO LOCK:

⬜ ONE SD1.5 anime checkpoint
⬜ ControlNet model(s)
⬜ F5-TTS model
⬜ LivePortrait model

IMPORTANT:
Do not start production yet.

After we choose the base anime checkpoint, we can lock the
ControlNet + VAE + remaining model files around it instead of
guessing and creating another compatibility mess.
""")

print("=" * 70)
```

```python
# ================================================================
# 🎬 JULIUS CAESAR — FINAL MODEL LOCK
# ComfyUI 0.37.0 / SD1.5 pipeline
# ================================================================

import os
import sys
import hashlib
import shutil
import subprocess
from pathlib import Path
from urllib.request import urlopen, Request

COMFY = Path("/content/ComfyUI")
MODELS = COMFY / "models"

print("=" * 70)
print("🎬 JULIUS CAESAR — FINAL MODEL LOCK")
print("=" * 70)
print(f"ComfyUI: {COMFY}")

if not COMFY.exists():
    raise RuntimeError(f"ComfyUI not found: {COMFY}")

# ------------------------------------------------
# 1. MODEL DIRECTORIES
# ------------------------------------------------

directories = {
    "checkpoints": MODELS / "checkpoints",
    "vae": MODELS / "vae",
    "animatediff": MODELS / "animatediff_models",
    "ipadapter": MODELS / "ipadapter",
    "clip_vision": MODELS / "clip_vision",
    "controlnet": MODELS / "controlnet",
    "loras": MODELS / "loras",
    "liveportrait": MODELS / "liveportrait",
}

print("\n📁 MODEL DIRECTORIES")

for name, path in directories.items():
    path.mkdir(parents=True, exist_ok=True)
    print(f"✅ {name}: {path}")

# ------------------------------------------------
# 2. DOWNLOAD HELPER
# ------------------------------------------------

def download_file(url, destination, expected_sha256=None):
    destination = Path(destination)

    if destination.exists():
        print(f"⏭️ Already exists: {destination.name}")

        if expected_sha256:
            print("   Checking SHA256...")
            sha = hashlib.sha256()
            with open(destination, "rb") as f:
                for chunk in iter(lambda: f.read(1024 * 1024), b""):
                    sha.update(chunk)

            actual = sha.hexdigest()

            if actual.lower() == expected_sha256.lower():
                print("   ✅ SHA256 verified")
                return True
            else:
                print("   ❌ SHA256 mismatch")
                print("   Removing corrupted file...")
                destination.unlink()

    print(f"\n⬇️ Downloading:")
    print(f"   {destination.name}")

    tmp = destination.with_suffix(destination.suffix + ".part")

    try:
        req = Request(
            url,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        with urlopen(req) as response, open(tmp, "wb") as out:
            total = response.headers.get("Content-Length")
            total = int(total) if total else None
            downloaded = 0

            while True:
                chunk = response.read(8 * 1024 * 1024)

                if not chunk:
                    break

                out.write(chunk)
                downloaded += len(chunk)

                if total:
                    pct = downloaded / total * 100
                    print(
                        f"\r   {downloaded / 1024**3:.2f} GB "
                        f"/ {total / 1024**3:.2f} GB "
                        f"({pct:.1f}%)",
                        end=""
                    )

        print()

        tmp.replace(destination)

        if expected_sha256:
            print("🔐 Verifying SHA256...")

            sha = hashlib.sha256()

            with open(destination, "rb") as f:
                for chunk in iter(lambda: f.read(1024 * 1024), b""):
                    sha.update(chunk)

            actual = sha.hexdigest()

            if actual.lower() != expected_sha256.lower():
                destination.unlink()
                raise RuntimeError(
                    f"SHA256 mismatch for {destination.name}\n"
                    f"Expected: {expected_sha256}\n"
                    f"Actual:   {actual}"
                )

            print("✅ SHA256 verified")

        print(f"✅ Download complete: {destination}")
        return True

    except Exception as e:
        if tmp.exists():
            tmp.unlink()

        print(f"❌ Download failed: {e}")
        return False


# ================================================================
# 3. BASE SD1.5 ANIME CHECKPOINT
# ================================================================

print("\n" + "=" * 70)
print("🎨 1/5 — BASE SD1.5 ANIME CHECKPOINT")
print("=" * 70)

checkpoint = directories["checkpoints"] / "Counterfeit-V3.0_fp16.safetensors"

counterfeit_url = (
    "https://huggingface.co/gsdf/Counterfeit-V3.0/"
    "resolve/main/Counterfeit-V3.0_fp16.safetensors"
)

counterfeit_sha256 = (
    "cbfba64e662370f59d4aa2aa69bf16749fce93846ccce20506aee5df01169859"
)

download_file(
    counterfeit_url,
    checkpoint,
    counterfeit_sha256
)

# ================================================================
# 4. CONTROLNET WEIGHTS
# ================================================================

print("\n" + "=" * 70)
print("🎛️ 2/5 — CONTROLNET")
print("=" * 70)

print("""
ControlNet extension is already installed.

For the final Caesar workflow we want:
  • OpenPose      → character/body posing
  • Lineart Anime → anime structure/line control
  • Depth         → scene geometry/composition

IMPORTANT:
The exact ControlNet filenames/repositories can vary between
SD1.5 model releases and workflow conventions.

Therefore this cell does NOT blindly download guessed ControlNet
weights.

The ControlNet directory is prepared and the workflow can lock
the exact compatible files once the workflow nodes are assembled.
""")

print(f"📁 ControlNet directory ready: {directories['controlnet']}")

# ================================================================
# 5. ANIMATEDIFF
# ================================================================

print("\n" + "=" * 70)
print("🎞️ 3/5 — ANIMATEDIFF")
print("=" * 70)

animatediff = directories["animatediff"] / "mm_sd_v15_v2.ckpt"

if animatediff.exists():
    size = animatediff.stat().st_size / 1024**3
    print(f"✅ AnimateDiff motion model found")
    print(f"   {animatediff}")
    print(f"   {size:.2f} GB")
else:
    print("❌ AnimateDiff motion model missing")
    print(f"   Expected: {animatediff}")

# ================================================================
# 6. IPADAPTER
# ================================================================

print("\n" + "=" * 70)
print("🧑 4/5 — IPADAPTER")
print("=" * 70)

ipadapter_files = [
    directories["ipadapter"] / "ip-adapter-plus_sd15.safetensors",
    directories["ipadapter"] / "ip-adapter-plus-face_sd15.safetensors",
]

clipvision = (
    directories["clip_vision"]
    / "CLIP-ViT-H-14-laion2B-s32B-b79K.safetensors"
)

for f in ipadapter_files:
    if f.exists():
        print(f"✅ {f.name}")
    else:
        print(f"❌ MISSING: {f}")

if clipvision.exists():
    print(f"✅ {clipvision.name}")
else:
    print(f"❌ MISSING: {clipvision}")

# ================================================================
# 7. LIVEPORTRAIT
# ================================================================

print("\n" + "=" * 70)
print("🙂 5/5 — LIVEPORTRAIT")
print("=" * 70)

liveportrait_node = (
    COMFY / "custom_nodes" / "ComfyUI-LivePortraitKJ"
)

if liveportrait_node.exists():
    print("✅ LivePortrait custom node already installed")
    print(f"   {liveportrait_node}")
else:
    print("⚠️ LivePortrait node not found.")
    print("""
Install the KJ LivePortrait node before using LivePortrait.

Its current implementation can automatically download the
compatible safetensor model set into:

    models/liveportrait/
""")

liveportrait_dir = directories["liveportrait"]

print(f"\n📁 LivePortrait model directory:")
print(f"   {liveportrait_dir}")

# ================================================================
# 8. VAE — DO NOT BLINDLY DUPLICATE
# ================================================================

print("\n" + "=" * 70)
print("🧩 VAE")
print("=" * 70)

vae_files = list(directories["vae"].glob("*"))

if vae_files:
    print("Existing VAE files:")
    for f in vae_files:
        if f.is_file():
            print(f"   • {f.name}")
else:
    print("""
ℹ️ No standalone VAE installed.

This is intentional.

We will use the VAE associated with the selected SD1.5
checkpoint/workflow rather than downloading an arbitrary VAE
and potentially creating a compatibility problem.
""")

# ================================================================
# 9. F5-TTS
# ================================================================

print("\n" + "=" * 70)
print("🗣️ F5-TTS")
print("=" * 70)

print("""
F5-TTS remains part of the FINAL production stack because Caesar
needs dialogue.

The actual F5-TTS model will be installed through the F5-TTS
ComfyUI node/model mechanism rather than guessing a model filename
here.

This avoids locking the wrong checkpoint or text encoder before
the dialogue workflow is assembled.
""")

# ================================================================
# 10. TEACACHE
# ================================================================

print("\n" + "=" * 70)
print("⚡ TEACACHE")
print("=" * 70)

print("""
⏭️ NOT INSTALLED.

Reason:
TeaCache is not a universal AnimateDiff/SD1.5 acceleration switch.

Our primary Caesar video pipeline is:
SD1.5 + AnimateDiff.

We will optimize that pipeline through ComfyUI's VRAM/offloading,
context-window and frame settings instead of introducing an
unverified acceleration dependency.
""")

# ================================================================
# 11. FINAL INVENTORY
# ================================================================

print("\n" + "=" * 70)
print("🔍 FINAL MODEL INVENTORY")
print("=" * 70)

checks = {
    "Base SD1.5 checkpoint":
        checkpoint,
    "AnimateDiff motion model":
        animatediff,
    "IPAdapter Plus":
        ipadapter_files[0],
    "IPAdapter Plus Face":
        ipadapter_files[1],
    "CLIP Vision ViT-H":
        clipvision,
}

for name, path in checks.items():
    if path.exists():
        size = path.stat().st_size / 1024**3
        print(f"✅ {name:<30} {size:.2f} GB")
    else:
        print(f"❌ {name:<30} MISSING")

print("\n" + "=" * 70)
print("🎬 CAESAR MODEL LOCK STATUS")
print("=" * 70)

required_now = [
    checkpoint,
    animatediff,
    ipadapter_files[0],
    ipadapter_files[1],
    clipvision,
]

if all(p.exists() for p in required_now):
    print("✅ CORE SD1.5 + ANIMATEDIFF + IPADAPTER STACK READY")
else:
    print("⚠️ CORE STACK IS NOT COMPLETE")

print("""
NEXT PRODUCTION COMPONENTS:

✅ SD1.5 anime base
✅ AnimateDiff
✅ IPAdapter Plus
✅ IPAdapter Face
✅ CLIP Vision
✅ ControlNet infrastructure
✅ Impact Pack infrastructure
✅ VHS infrastructure

🔜 ControlNet workflow-specific weights
🔜 LivePortrait node + weights
🔜 F5-TTS node + model
🔜 Final Caesar workflow JSON

🚫 TeaCache intentionally excluded from the SD1.5 AnimateDiff stack.

DO NOT START GENERATING SCENES YET.
The next step is building the actual Caesar workflow around this
locked core rather than downloading more random weights.
""")

print("=" * 70)
print("🏛️ JULIUS CAESAR MODEL INFRASTRUCTURE CHECK COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# 🎬 JULIUS CAESAR — MASTER COMPONENT INSTALLER
# ControlNet (Fixed URLs) + LivePortrait + F5-TTS
# ================================================================

import os
import sys
import subprocess
import hashlib
from pathlib import Path
from urllib.request import urlopen, Request

COMFY = Path("/content/ComfyUI")
CUSTOM = COMFY / "custom_nodes"
MODELS = COMFY / "models"

CONTROLNET = MODELS / "controlnet"
LIVEPORTRAIT = MODELS / "liveportrait"

CONTROLNET.mkdir(parents=True, exist_ok=True)
LIVEPORTRAIT.mkdir(parents=True, exist_ok=True)
CUSTOM.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("🎬 JULIUS CAESAR — FINAL PRODUCTION COMPONENT INSTALL")
print("=" * 70)


# ================================================================
# HELPERS
# ================================================================

def run(cmd, cwd=None):
    print("\n$", " ".join(cmd))
    result = subprocess.run(cmd, cwd=cwd, text=True)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with return code {result.returncode}")
    return result

def clone_or_update(url, folder):
    folder = Path(folder)
    if folder.exists() and (folder / ".git").exists():
        print(f"⏭️ Already cloned: {folder.name}")
        print("   Pulling latest changes...")
        run(["git", "-C", str(folder), "pull", "--ff-only"])
    else:
        print(f"⬇️ Cloning: {folder.name}")
        run(["git", "clone", "--depth", "1", url, str(folder)])
        print("✅ Clone complete")

def download_verified(filename, url, expected_sha256=None):
    dest = CONTROLNET / filename
    if dest.exists():
        size = dest.stat().st_size / 1024**3
        print(f"⏭️ Already exists: {filename} ({size:.2f} GB)")
        return

    print(f"\n⬇️ Downloading {filename}...")
    tmp = dest.with_suffix(dest.suffix + ".part")
    req = Request(url, headers={"User-Agent": "Mozilla/5.0"})

    with urlopen(req) as response, open(tmp, "wb") as f:
        total = int(response.headers.get("Content-Length", 0))
        done = 0
        while True:
            chunk = response.read(8 * 1024 * 1024)
            if not chunk:
                break
            f.write(chunk)
            done += len(chunk)
            if total:
                print(f"\r   {done/1024**3:.2f} / {total/1024**3:.2f} GB ({done/total*100:.1f}%)", end="")
    print()
    tmp.replace(dest)

    if expected_sha256:
        print("🔐 Checking SHA256...")
        h = hashlib.sha256()
        with open(dest, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)
        actual = h.hexdigest()
        if actual.lower() != expected_sha256.lower():
            dest.unlink(missing_ok=True)
            raise RuntimeError(f"SHA256 mismatch for {filename}\nExpected: {expected_sha256}\nActual:   {actual}")
        print(f"✅ VERIFIED: {filename}")
    else:
        print(f"✅ DOWNLOADED: {filename}")


# ================================================================
# 1. CONTROLNET WEIGHTS (CORRECTED URLS)
# ================================================================

print("\n" + "=" * 70)
print("🎛️ 1/3 — SD1.5 CONTROLNET WEIGHTS")
print("=" * 70)

controlnets = [
    {
        "file": "control_v11p_sd15_openpose.pth",
        "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11p_sd15_openpose.pth",
        "hash": "db97becd92cd19aff71352a60e93c2508decba3dee64f01f686727b9b406a9dd"
    },
    {
        "file": "control_v11f1p_sd15_depth.pth",
        "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11f1p_sd15_depth.pth",
        "hash": None
    },
    {
        "file": "control_v11p_sd15s2_lineart_anime.pth",
        "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11p_sd15s2_lineart_anime.pth",
        "hash": None
    }
]

for cn in controlnets:
    download_verified(cn["file"], cn["url"], cn["hash"])


# ================================================================
# 2. LIVEPORTRAIT
# ================================================================

print("\n" + "=" * 70)
print("🙂 2/3 — LIVEPORTRAIT")
print("=" * 70)

liveportrait_node = CUSTOM / "ComfyUI-LivePortraitKJ"
clone_or_update("https://github.com/kijai/ComfyUI-LivePortraitKJ.git", liveportrait_node)

req_lp = liveportrait_node / "requirements.txt"
if req_lp.exists():
    print("\n📚 Installing LivePortrait requirements...")
    run([sys.executable, "-m", "pip", "install", "-r", str(req_lp)])


# ================================================================
# 3. F5-TTS
# ================================================================

print("\n" + "=" * 70)
print("🗣️ 3/3 — F5-TTS")
print("=" * 70)

f5_node = CUSTOM / "ComfyUI-F5-TTS"
clone_or_update("https://github.com/niknah/ComfyUI-F5-TTS.git", f5_node)

print("\n📦 Initializing F5-TTS submodules...")
run(["git", "submodule", "update", "--init", "--recursive"], cwd=str(f5_node))

req_f5 = f5_node / "requirements.txt"
if req_f5.exists():
    print("\n📚 Installing F5-TTS requirements...")
    run([sys.executable, "-m", "pip", "install", "-r", str(req_f5)])

print("\n" + "=" * 70)
print("✅ ALL COMPONENTS SUCCESSFULLY INSTALLED.")
print("   Restart ComfyUI and refresh the browser.")
print("=" * 70)
```

```python
# ============================================================
# 🟡 LOCALTUNNEL — RUN LAST
# ============================================================

import subprocess
import time
import re

PORT = "8188"

print("=" * 65)
print("🟡 STARTING LOCALTUNNEL")
print("=" * 65)

# Make sure no previous LocalTunnel process is hanging around
subprocess.run(
    "pkill -f 'localtunnel.*8188' || true",
    shell=True
)

time.sleep(2)

print("Connecting to ComfyUI on port 8188...")

tunnel = subprocess.Popen(
    [
        "npx",
        "--yes",
        "localtunnel",
        "--port",
        PORT
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

url = None

start = time.time()

while time.time() - start < 30:
    line = tunnel.stdout.readline()

    if line:
        print(line.strip())

        match = re.search(
            r"https?://[^\s]+",
            line
        )

        if match:
            url = match.group(0)
            break

    time.sleep(0.2)

print("\n" + "=" * 65)

if url:
    print("✅ LOCALTUNNEL READY")
    print("\n🌐 YOUR COMFYUI URL:")
    print(url)
    print("\nOpen that URL in your browser.")
else:
    print("❌ LocalTunnel URL was not detected.")
    print("Tunnel process is still running.")
    print("PID:", tunnel.pid)

print("=" * 65)
```
