```python
# ============================================================
# 🚀 COMFYUI MASTER STARTUP
# DagsHub → Fast Storage → Dependencies → Manager
# ============================================================

import os
import sys
import shutil
import subprocess
import time
from pathlib import Path

COMFY = Path("/content/ComfyUI")
DAG_COMFY = Path("/content/dagshub_storage/comfyui_project/ComfyUI")
MOUNT = Path("/content/dagshub_storage")
REPO = "dpbasvs-afk/comfyui-project"

print("=" * 65)
print("🚀 COMFYUI MASTER STARTUP")
print("=" * 65)

# ------------------------------------------------------------
# 1. DagsHub
# ------------------------------------------------------------
print("\n[1/5] Checking DagsHub...")

try:
    import dagshub
    print("DagsHub package: OK")

    # Check whether existing mount works
    if not MOUNT.exists():
        print("Mount missing — connecting DagsHub...")
        dagshub.storage.mount(REPO, path=str(MOUNT))
    else:
        try:
            list(MOUNT.iterdir())
            print("DagsHub mount: OK")
        except OSError:
            print("Broken mount detected — reconnecting...")
            try:
                dagshub.storage.unmount(
                    repo=REPO,
                    path=str(MOUNT)
                )
            except Exception:
                pass

            time.sleep(2)

            dagshub.storage.mount(
                REPO,
                path=str(MOUNT)
            )

    print("DagsHub: READY")

except Exception as e:
    print("❌ DagsHub startup failed:")
    print(e)
    raise

# ------------------------------------------------------------
# 2. Verify source
# ------------------------------------------------------------
print("\n[2/5] Checking DagsHub ComfyUI...")

if not DAG_COMFY.exists():
    raise RuntimeError(f"ComfyUI not found at {DAG_COMFY}")

print("Source:", DAG_COMFY)

required = [
    "main.py",
    "comfy",
    "comfy_api",
    "comfy_api/input",
    "comfy_api/latest",
    "comfy_api/latest/_io.py",
]

for item in required:
    p = DAG_COMFY / item
    print(("✅ " if p.exists() else "❌ ") + item)

# ------------------------------------------------------------
# 3. Copy to FAST Colab storage
# ------------------------------------------------------------
print("\n[3/5] Preparing FAST ComfyUI...")

if COMFY.exists():
    print("Fast ComfyUI already exists.")

    # Verify it isn't an incomplete/broken copy
    if not (COMFY / "main.py").exists():
        print("Incomplete Fast copy detected — rebuilding...")
        shutil.rmtree(COMFY, ignore_errors=True)

if not COMFY.exists():
    print("Copying ComfyUI from DagsHub to fast storage...")
    shutil.copytree(DAG_COMFY, COMFY)
    print("✅ Fast copy completed")

print("Fast ComfyUI:", COMFY)

# ------------------------------------------------------------
# 4. Python dependencies
# ------------------------------------------------------------
print("\n[4/5] Checking Python dependencies...")

REQ = COMFY / "requirements.txt"

if REQ.exists():
    print("Installing/refreshing ComfyUI requirements...")
    result = subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-r",
            str(REQ),
            "-q"
        ]
    )

    if result.returncode != 0:
        raise RuntimeError("ComfyUI dependency installation failed.")

print("✅ Dependencies ready")

# ------------------------------------------------------------
# 5. ComfyUI Manager 4.2.2
# ------------------------------------------------------------
print("\n[5/5] Checking ComfyUI Manager...")

MANAGER_REQ = COMFY / "manager_requirements.txt"

if MANAGER_REQ.exists():
    print("Installing Manager requirement...")
    result = subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-r",
            str(MANAGER_REQ),
            "-q"
        ]
    )

    if result.returncode != 0:
        raise RuntimeError("Manager installation failed.")

print("Testing Manager import...")

try:
    import comfyui_manager
    print("✅ comfyui_manager available")
except Exception as e:
    raise RuntimeError(
        f"Manager import failed: {e}"
    )

# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------
print("\n" + "=" * 65)
print("✅ MASTER STARTUP COMPLETE")
print("=" * 65)

print("ComfyUI:", COMFY)
print("main.py:", (COMFY / "main.py").exists())
print("comfy_api:", (COMFY / "comfy_api").exists())
print("Manager: READY")

print("\nNEXT CELL:")
print("👉 Start ComfyUI + Manager")
print("👉 Do NOT start LocalTunnel yet")
print("=" * 65)
```

```python
# ============================================================
# 🟦 START COMFYUI + MANAGER
# ============================================================

import os
import sys
import subprocess
import time

COMFY = "/content/ComfyUI"

os.chdir(COMFY)

print("=" * 65)
print("🟦 STARTING COMFYUI + MANAGER")
print("=" * 65)

# Stop ALL previous ComfyUI instances
print("\nStopping old ComfyUI processes...")

subprocess.run(
    "pkill -f '/content/ComfyUI/main.py' || true",
    shell=True
)

time.sleep(3)

# Start fresh instance
print("Starting fresh ComfyUI...")
print("Manager: ENABLED")
print("Port: 8188")

process = subprocess.Popen(
    [
        sys.executable,
        "main.py",
        "--listen", "0.0.0.0",
        "--port", "8188",
        "--cuda-device", "0",
        "--enable-manager"
    ],
    cwd=COMFY
)

print("PID:", process.pid)

# Give ComfyUI time to initialize
print("\nWaiting for server...")

ready = False

for i in range(30):
    time.sleep(1)

    result = subprocess.run(
        ["curl", "-s", "-o", "/dev/null", "-w", "%{http_code}", "http://127.0.0.1:8188"],
        capture_output=True,
        text=True
    )

    if result.stdout in ["200", "302", "404"]:
        ready = True
        break

if ready:
    print(f"\n✅ COMFYUI READY after ~{i+1} seconds")
else:
    print("\n⚠️ Server did not respond within 30 seconds.")
    print("Check the ComfyUI output above.")

# Verify Manager flag
check = subprocess.run(
    ["pgrep", "-af", "main.py"],
    capture_output=True,
    text=True
)

print("\n========== PROCESS CHECK ==========")
print(check.stdout)

if "--enable-manager" in check.stdout:
    print("✅ MANAGER ENABLED")
else:
    print("❌ MANAGER FLAG NOT FOUND")

print("\n" + "=" * 65)
print("COMFYUI START CELL FINISHED")
print("=" * 65)
print("👉 NOW run the LocalTunnel cell.")
```

```python
# ============================================================
# 🚀 RESTART COMFYUI WITH MANAGER — SINGLE PROCESS
# ============================================================

import os
import signal
import subprocess
import time
import requests

COMFY = "/content/ComfyUI"
PORT = 8188

print("=" * 65)
print("🚀 STARTING COMFYUI 0.37.0 + MANAGER 4.2.2")
print("=" * 65)

# ------------------------------------------------------------
# 1. Kill EVERY old ComfyUI process
# ------------------------------------------------------------

print("\n[1/4] Cleaning old ComfyUI processes...")

subprocess.run(
    "pkill -f 'python.*main.py.*--port 8188' || true",
    shell=True
)

time.sleep(3)

# ------------------------------------------------------------
# 2. Verify port is free
# ------------------------------------------------------------

print("[2/4] Checking port 8188...")

check = subprocess.run(
    "ps aux | grep '[m]ain.py.*8188'",
    shell=True,
    capture_output=True,
    text=True
)

if check.stdout.strip():
    print("⚠️ Old process still detected:")
    print(check.stdout)
    raise RuntimeError("Old ComfyUI process is still running.")

print("✅ Port/process cleanup complete")

# ------------------------------------------------------------
# 3. Start ComfyUI WITH MANAGER
# ------------------------------------------------------------

print("\n[3/4] Starting ComfyUI + Manager...")

log_file = open("/content/comfyui_manager.log", "w")

process = subprocess.Popen(
    [
        "python",
        "main.py",
        "--listen", "0.0.0.0",
        "--port", str(PORT),
        "--cuda-device", "0",
        "--enable-manager"
    ],
    cwd=COMFY,
    stdout=log_file,
    stderr=subprocess.STDOUT
)

print("PID:", process.pid)
print("Manager flag: --enable-manager")

# ------------------------------------------------------------
# 4. Wait for server
# ------------------------------------------------------------

print("\n[4/4] Waiting for ComfyUI...")

ready = False

for i in range(30):
    time.sleep(1)

    try:
        r = requests.get(
            "http://127.0.0.1:8188/system_stats",
            timeout=2
        )

        if r.status_code == 200:
            ready = True
            break

    except Exception:
        pass

if not ready:
    print("\n❌ ComfyUI did not become ready.")
    print("\nLast log output:\n")

    try:
        with open("/content/comfyui_manager.log", "r") as f:
            print(f.read()[-10000:])
    except:
        pass

    raise RuntimeError("ComfyUI startup failed.")

# ------------------------------------------------------------
# VERIFY ACTUAL ARGUMENTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("🔍 VERIFYING RUNNING PROCESS")
print("=" * 65)

ps = subprocess.run(
    f"ps -p {process.pid} -o pid,args=",
    shell=True,
    capture_output=True,
    text=True
)

print(ps.stdout)

if "--enable-manager" not in ps.stdout:
    raise RuntimeError(
        "❌ Manager flag is NOT present in the running process."
    )

print("✅ --enable-manager CONFIRMED")

# ------------------------------------------------------------
# VERIFY SERVER FEATURES
# ------------------------------------------------------------

features = requests.get(
    "http://127.0.0.1:8188/api/features",
    timeout=5
).json()

print("\nManager feature response:")
print(features.get("extension", {}).get("manager"))

print("\n" + "=" * 65)
print("✅ COMFYUI + MANAGER IS ACTUALLY RUNNING")
print("=" * 65)

print("\nPID:", process.pid)
print("Local server: http://127.0.0.1:8188")
print("\n⚠️ DO NOT START THE TUNNEL YET.")
print("After this cell succeeds, open the tunnel separately.")
```

```python
# ================================================================
# 🚀 JULIUS CAESAR — PERMANENT EXTENSION SETUP
# ================================================================
# RUN ORDER:
#   1. MASTER STARTUP CELL
#   2. THIS CELL  ← extensions/dependencies
#   3. COMFYUI + MANAGER START CELL (if your startup doesn't do it)
#   4. TUNNEL CELL
#
# This cell is SAFE TO RUN AGAIN.
# Existing repositories are NOT recloned.
# ================================================================

import os
import sys
import subprocess
from pathlib import Path

print("=" * 70)
print("🎬 JULIUS CAESAR — EXTENSION SETUP")
print("=" * 70)

# ------------------------------------------------
# 1. LOCATIONS
# ------------------------------------------------

COMFY = Path("/content/ComfyUI")
CUSTOM = COMFY / "custom_nodes"

if not COMFY.exists():
    raise RuntimeError(f"❌ ComfyUI not found: {COMFY}")

CUSTOM.mkdir(parents=True, exist_ok=True)

print(f"\nComfyUI : {COMFY}")
print(f"Custom nodes : {CUSTOM}")

# ------------------------------------------------
# 2. REPOSITORIES
# ------------------------------------------------
# Official repositories for the production stack.
#
# We are installing:
#   AnimateDiff Evolved
#   VideoHelperSuite
#   IPAdapter Plus
#   Advanced ControlNet
#   ControlNet Aux
#   Impact Pack
#   TeaCache
#
# F5-TTS and LivePortrait are handled separately because
# they are not simply ordinary ComfyUI custom-node folders.
# ------------------------------------------------

REPOS = {
    "ComfyUI-AnimateDiff-Evolved":
        "https://github.com/Kosinkadink/ComfyUI-AnimateDiff-Evolved.git",

    "ComfyUI-VideoHelperSuite":
        "https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite.git",

    "ComfyUI_IPAdapter_plus":
        "https://github.com/cubiq/ComfyUI_IPAdapter_plus.git",

    "ComfyUI-Advanced-ControlNet":
        "https://github.com/Kosinkadink/ComfyUI-Advanced-ControlNet.git",

    "comfyui_controlnet_aux":
        "https://github.com/Fannovel16/comfyui_controlnet_aux.git",

    "ComfyUI-Impact-Pack":
        "https://github.com/ltdrdata/ComfyUI-Impact-Pack.git",

    "ComfyUI-TeaCache":
        "https://github.com/nvmax/teacache.git",
}

# ------------------------------------------------
# 3. GIT INSTALLER
# ------------------------------------------------

def run_cmd(cmd, cwd=None):
    print("\n$", " ".join(map(str, cmd)))
    result = subprocess.run(
        cmd,
        cwd=str(cwd) if cwd else None,
        text=True
    )
    if result.returncode != 0:
        print(f"⚠️ Command returned {result.returncode}")
    return result.returncode


def install_or_update(name, url):

    target = CUSTOM / name

    print("\n" + "-" * 65)
    print(f"📦 {name}")
    print("-" * 65)

    if target.exists() and (target / ".git").exists():

        print("✅ Already cloned")

        # Do NOT blindly pull every time.
        # The current checkout is preserved to avoid unexpected
        # changes breaking a working production environment.

        print("ℹ️ Existing checkout preserved.")

    elif target.exists():

        print("⚠️ Folder exists but is not a Git repository.")
        print("Keeping it untouched:", target)

    else:

        print("⬇️ Cloning...")
        rc = run_cmd(
            ["git", "clone", "--depth", "1", url, str(target)]
        )

        if rc == 0:
            print("✅ Clone complete")
        else:
            print("❌ Clone failed")

    # ------------------------------------------------
    # Install extension-specific Python requirements
    # ------------------------------------------------

    requirements = target / "requirements.txt"

    if requirements.exists():

        print(f"📚 Installing requirements for {name}...")

        rc = run_cmd(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-r",
                str(requirements)
            ]
        )

        if rc == 0:
            print("✅ Requirements OK")
        else:
            print("⚠️ Some requirements may have failed.")

    else:
        print("ℹ️ No requirements.txt")

# ------------------------------------------------
# 4. INSTALL EVERYTHING
# ------------------------------------------------

for name, url in REPOS.items():
    install_or_update(name, url)

# ------------------------------------------------
# 5. MOVIEPY
# ------------------------------------------------

print("\n" + "=" * 65)
print("🎞️ MOVIEPY")
print("=" * 65)

run_cmd([
    sys.executable,
    "-m",
    "pip",
    "install",
    "moviepy"
])

print("✅ MoviePy installation attempted")

# ------------------------------------------------
# 6. VERIFY EXTENSIONS
# ------------------------------------------------

print("\n" + "=" * 70)
print("🔍 VERIFYING EXTENSIONS")
print("=" * 70)

for name in REPOS:
    path = CUSTOM / name

    if path.exists():
        print(f"✅ {name}")
    else:
        print(f"❌ {name}")

# ------------------------------------------------
# 7. VERIFY IMPORTANT COMFYUI DIRECTORIES
# ------------------------------------------------

print("\n" + "=" * 70)
print("📁 VERIFYING MODEL DIRECTORIES")
print("=" * 70)

MODEL_DIRS = [
    COMFY / "models" / "checkpoints",
    COMFY / "models" / "controlnet",
    COMFY / "models" / "clip_vision",
    COMFY / "models" / "animatediff_models",
    COMFY / "models" / "ipadapter",
    COMFY / "models" / "vae",
]

for path in MODEL_DIRS:
    path.mkdir(parents=True, exist_ok=True)
    print(f"✅ {path}")

# ------------------------------------------------
# 8. FINAL STATUS
# ------------------------------------------------

print("\n" + "=" * 70)
print("🎬 JULIUS CAESAR EXTENSION SETUP COMPLETE")
print("=" * 70)

print("""
Installed / prepared:

✅ AnimateDiff Evolved
✅ VideoHelperSuite
✅ IPAdapter Plus
✅ Advanced ControlNet
✅ ControlNet Aux
✅ Impact Pack
✅ TeaCache
✅ MoviePy

Prepared model folders:

✅ checkpoints
✅ controlnet
✅ clip_vision
✅ animatediff_models
✅ ipadapter
✅ vae

NOT downloaded automatically yet:

⬇️ Anime checkpoint
⬇️ AnimateDiff motion model
⬇️ IPAdapter model
⬇️ CLIP Vision model
⬇️ ControlNet model(s)
⬇️ F5-TTS model
⬇️ LivePortrait model

Those are MODEL WEIGHTS, not extensions.

IMPORTANT:
Restart ComfyUI after installing/updating custom nodes.
Do not start the tunnel until ComfyUI has restarted successfully.
""")

print("=" * 70)
print("✅ ONE-CELL EXTENSION SETUP FINISHED")
print("=" * 70)
```

```python
# ============================================================
# 🟡 LOCALTUNNEL — RUN LAST
# ============================================================

import subprocess
import time
import re

PORT = "8188"

print("=" * 65)
print("🟡 STARTING LOCALTUNNEL")
print("=" * 65)

# Make sure no previous LocalTunnel process is hanging around
subprocess.run(
    "pkill -f 'localtunnel.*8188' || true",
    shell=True
)

time.sleep(2)

print("Connecting to ComfyUI on port 8188...")

tunnel = subprocess.Popen(
    [
        "npx",
        "--yes",
        "localtunnel",
        "--port",
        PORT
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

url = None

start = time.time()

while time.time() - start < 30:
    line = tunnel.stdout.readline()

    if line:
        print(line.strip())

        match = re.search(
            r"https?://[^\s]+",
            line
        )

        if match:
            url = match.group(0)
            break

    time.sleep(0.2)

print("\n" + "=" * 65)

if url:
    print("✅ LOCALTUNNEL READY")
    print("\n🌐 YOUR COMFYUI URL:")
    print(url)
    print("\nOpen that URL in your browser.")
else:
    print("❌ LocalTunnel URL was not detected.")
    print("Tunnel process is still running.")
    print("PID:", tunnel.pid)

print("=" * 65)
```
