```python
# ======================================================================
# 🎬 JULIUS CAESAR — MODEL WEIGHTS SETUP
# ======================================================================
# RUN AFTER:
#   1. MASTER STARTUP
#   2. EXTENSION SETUP
#
# This cell is SAFE TO RUN AGAIN.
# Existing files are NOT downloaded again.
#
# CURRENT STACK:
#   SD 1.5
#   AnimateDiff Evolved
#   IPAdapter Plus
#   CLIP Vision ViT-H
#
# ======================================================================

import os
import sys
import hashlib
import urllib.request
from pathlib import Path

print("=" * 70)
print("🎬 JULIUS CAESAR — MODEL WEIGHTS SETUP")
print("=" * 70)

# ----------------------------------------------------------------------
# COMFYUI
# ----------------------------------------------------------------------

COMFY = Path("/content/ComfyUI")

if not COMFY.exists():
    raise RuntimeError(f"❌ ComfyUI not found: {COMFY}")

print(f"ComfyUI: {COMFY}")

# ----------------------------------------------------------------------
# MODEL DIRECTORIES
# ----------------------------------------------------------------------

DIRS = {
    "checkpoints": COMFY / "models" / "checkpoints",
    "vae": COMFY / "models" / "vae",
    "animatediff": COMFY / "models" / "animatediff_models",
    "ipadapter": COMFY / "models" / "ipadapter",
    "clip_vision": COMFY / "models" / "clip_vision",
    "controlnet": COMFY / "models" / "controlnet",
    "loras": COMFY / "models" / "loras",
}

print("\n📁 Creating model directories...")

for name, path in DIRS.items():
    path.mkdir(parents=True, exist_ok=True)
    print(f"✅ {name}: {path}")

# ----------------------------------------------------------------------
# DOWNLOAD FUNCTION
# ----------------------------------------------------------------------

def download_if_missing(url, destination, expected_sha256=None):

    destination = Path(destination)

    if destination.exists():

        size_mb = destination.stat().st_size / (1024 * 1024)

        if size_mb > 1:
            print(f"✅ Already exists: {destination.name} ({size_mb:.1f} MB)")
            return True

        print(f"⚠️ Existing file looks incomplete: {destination.name}")
        print("Removing incomplete file...")
        destination.unlink()

    print("\n⬇️ Downloading:")
    print(destination.name)
    print("From:", url)

    try:

        urllib.request.urlretrieve(url, destination)

    except Exception as e:

        print("❌ Download failed:", e)

        if destination.exists():
            destination.unlink()

        return False

    if not destination.exists():
        print("❌ File was not created.")
        return False

    size_mb = destination.stat().st_size / (1024 * 1024)

    print(f"✅ Downloaded: {size_mb:.1f} MB")

    # --------------------------------------------------------------
    # Optional SHA256 verification
    # --------------------------------------------------------------

    if expected_sha256:

        print("🔐 Verifying SHA256...")

        sha = hashlib.sha256()

        with open(destination, "rb") as f:
            while True:
                chunk = f.read(1024 * 1024)

                if not chunk:
                    break

                sha.update(chunk)

        actual = sha.hexdigest()

        if actual.lower() != expected_sha256.lower():

            print("❌ SHA256 MISMATCH")
            print("Expected:", expected_sha256)
            print("Actual:  ", actual)

            destination.unlink()

            return False

        print("✅ SHA256 verified")

    return True


# ======================================================================
# 1. ANIMATEDIFF MOTION MODEL
# ======================================================================

print("\n" + "=" * 70)
print("🎞️ 1/4 — ANIMATEDIFF MOTION MODEL")
print("=" * 70)

ANIMATEDIFF_URL = (
    "https://huggingface.co/guoyww/animatediff/resolve/main/"
    "mm_sd_v15_v2.safetensors"
)

ANIMATEDIFF_FILE = DIRS["animatediff"] / "mm_sd_v15_v2.safetensors"

download_if_missing(
    ANIMATEDIFF_URL,
    ANIMATEDIFF_FILE,
    expected_sha256=(
        "1cf5d6d6062b8d3d72cd5cdcabb578befd79b7a3e1ea41bf15d3ca594c0998da"
    )
)

# This motion module is 1.82 GB according to its Hugging Face listing.
# It is specifically an SD1.5 AnimateDiff motion module.
# ======================================================================


# ======================================================================
# 2. IPADAPTER PLUS SD1.5
# ======================================================================

print("\n" + "=" * 70)
print("🧑 2/4 — IPADAPTER PLUS SD1.5")
print("=" * 70)

IPADAPTER_URL = (
    "https://huggingface.co/h94/IP-Adapter/resolve/main/"
    "models/ip-adapter-plus_sd15.safetensors"
)

IPADAPTER_FILE = DIRS["ipadapter"] / "ip-adapter-plus_sd15.safetensors"

download_if_missing(
    IPADAPTER_URL,
    IPADAPTER_FILE,
    expected_sha256=(
        "a1c250be40455cc61a43da1201ec3f1edaea71214865fb47f57927e06cbe4996"
    )
)

# ======================================================================
# 3. IPADAPTER PLUS FACE
# ======================================================================

print("\n" + "=" * 70)
print("🙂 3/4 — IPADAPTER PLUS FACE")
print("=" * 70)

IPADAPTER_FACE_URL = (
    "https://huggingface.co/h94/IP-Adapter/resolve/main/"
    "models/ip-adapter-plus-face_sd15.safetensors"
)

IPADAPTER_FACE_FILE = (
    DIRS["ipadapter"] /
    "ip-adapter-plus-face_sd15.safetensors"
)

download_if_missing(
    IPADAPTER_FACE_URL,
    IPADAPTER_FACE_FILE,
    expected_sha256=(
        "1c9edc21af6f737dc1d6e0e734190e976cfacf802d6b024b77aa3be922f7569b"
    )
)

# ======================================================================
# 4. CLIP VISION — ViT-H
# ======================================================================

print("\n" + "=" * 70)
print("👁️ 4/4 — CLIP VISION ViT-H")
print("=" * 70)

CLIP_VISION_URL = (
    "https://huggingface.co/h94/IP-Adapter/resolve/main/"
    "models/image_encoder/model.safetensors"
)

CLIP_VISION_FILE = (
    DIRS["clip_vision"] /
    "CLIP-ViT-H-14-laion2B-s32B-b79K.safetensors"
)

download_if_missing(
    CLIP_VISION_URL,
    CLIP_VISION_FILE
)

# ======================================================================
# CONTROLNET
# ======================================================================

print("\n" + "=" * 70)
print("🎯 CONTROLNET")
print("=" * 70)

print("""
ControlNet models are NOT being blindly downloaded yet.

Reason:
The exact ControlNet checkpoint must match the final SD1.5
anime checkpoint and the controls we actually use.

The ControlNet-Aux extension is already installed and can provide
the preprocessing nodes.

We will add only the ControlNet weights actually required by
the Caesar workflow.
""")

# ======================================================================
# BASE ANIME CHECKPOINT
# ======================================================================

print("=" * 70)
print("🎨 BASE ANIME CHECKPOINT")
print("=" * 70)

print("""
⚠️ NOT DOWNLOADED YET.

This is intentional.

We need to choose ONE exact SD1.5 anime checkpoint for the Caesar
visual style before locking the rest of the generation stack.

Do NOT put a random .safetensors file here.

Once the checkpoint is chosen, it will go into:

    /content/ComfyUI/models/checkpoints/

The rest of this cell is already prepared for it.
""")

# ======================================================================
# FINAL VERIFICATION
# ======================================================================

print("\n" + "=" * 70)
print("🔍 FINAL MODEL CHECK")
print("=" * 70)

def show_file(label, path):

    if path.exists():

        size_gb = path.stat().st_size / (1024 ** 3)

        print(f"✅ {label}")
        print(f"   {path}")
        print(f"   {size_gb:.2f} GB")

    else:

        print(f"❌ {label}")
        print(f"   Missing: {path}")


show_file(
    "AnimateDiff motion model",
    ANIMATEDIFF_FILE
)

show_file(
    "IPAdapter Plus",
    IPADAPTER_FILE
)

show_file(
    "IPAdapter Plus Face",
    IPADAPTER_FACE_FILE
)

show_file(
    "CLIP Vision ViT-H",
    CLIP_VISION_FILE
)

print("\n" + "=" * 70)
print("🎬 MODEL INFRASTRUCTURE SETUP COMPLETE")
print("=" * 70)

print("""
READY:

✅ AnimateDiff SD1.5 motion model
✅ IPAdapter Plus SD1.5
✅ IPAdapter Plus Face SD1.5
✅ CLIP Vision ViT-H
✅ Required model folders

STILL TO LOCK:

⬜ ONE SD1.5 anime checkpoint
⬜ ControlNet model(s)
⬜ F5-TTS model
⬜ LivePortrait model

IMPORTANT:
Do not start production yet.

After we choose the base anime checkpoint, we can lock the
ControlNet + VAE + remaining model files around it instead of
guessing and creating another compatibility mess.
""")

print("=" * 70)
```

```python
# ================================================================
# 🎬 JULIUS CAESAR — FINAL MODEL LOCK
# ComfyUI 0.37.0 / SD1.5 pipeline
# ================================================================

import os
import sys
import hashlib
import shutil
import subprocess
from pathlib import Path
from urllib.request import urlopen, Request

COMFY = Path("/content/ComfyUI")
MODELS = COMFY / "models"

print("=" * 70)
print("🎬 JULIUS CAESAR — FINAL MODEL LOCK")
print("=" * 70)
print(f"ComfyUI: {COMFY}")

if not COMFY.exists():
    raise RuntimeError(f"ComfyUI not found: {COMFY}")

# ------------------------------------------------
# 1. MODEL DIRECTORIES
# ------------------------------------------------

directories = {
    "checkpoints": MODELS / "checkpoints",
    "vae": MODELS / "vae",
    "animatediff": MODELS / "animatediff_models",
    "ipadapter": MODELS / "ipadapter",
    "clip_vision": MODELS / "clip_vision",
    "controlnet": MODELS / "controlnet",
    "loras": MODELS / "loras",
    "liveportrait": MODELS / "liveportrait",
}

print("\n📁 MODEL DIRECTORIES")

for name, path in directories.items():
    path.mkdir(parents=True, exist_ok=True)
    print(f"✅ {name}: {path}")

# ------------------------------------------------
# 2. DOWNLOAD HELPER
# ------------------------------------------------

def download_file(url, destination, expected_sha256=None):
    destination = Path(destination)

    if destination.exists():
        print(f"⏭️ Already exists: {destination.name}")

        if expected_sha256:
            print("   Checking SHA256...")
            sha = hashlib.sha256()
            with open(destination, "rb") as f:
                for chunk in iter(lambda: f.read(1024 * 1024), b""):
                    sha.update(chunk)

            actual = sha.hexdigest()

            if actual.lower() == expected_sha256.lower():
                print("   ✅ SHA256 verified")
                return True
            else:
                print("   ❌ SHA256 mismatch")
                print("   Removing corrupted file...")
                destination.unlink()

    print(f"\n⬇️ Downloading:")
    print(f"   {destination.name}")

    tmp = destination.with_suffix(destination.suffix + ".part")

    try:
        req = Request(
            url,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        with urlopen(req) as response, open(tmp, "wb") as out:
            total = response.headers.get("Content-Length")
            total = int(total) if total else None
            downloaded = 0

            while True:
                chunk = response.read(8 * 1024 * 1024)

                if not chunk:
                    break

                out.write(chunk)
                downloaded += len(chunk)

                if total:
                    pct = downloaded / total * 100
                    print(
                        f"\r   {downloaded / 1024**3:.2f} GB "
                        f"/ {total / 1024**3:.2f} GB "
                        f"({pct:.1f}%)",
                        end=""
                    )

        print()

        tmp.replace(destination)

        if expected_sha256:
            print("🔐 Verifying SHA256...")

            sha = hashlib.sha256()

            with open(destination, "rb") as f:
                for chunk in iter(lambda: f.read(1024 * 1024), b""):
                    sha.update(chunk)

            actual = sha.hexdigest()

            if actual.lower() != expected_sha256.lower():
                destination.unlink()
                raise RuntimeError(
                    f"SHA256 mismatch for {destination.name}\n"
                    f"Expected: {expected_sha256}\n"
                    f"Actual:   {actual}"
                )

            print("✅ SHA256 verified")

        print(f"✅ Download complete: {destination}")
        return True

    except Exception as e:
        if tmp.exists():
            tmp.unlink()

        print(f"❌ Download failed: {e}")
        return False


# ================================================================
# 3. BASE SD1.5 ANIME CHECKPOINT
# ================================================================

print("\n" + "=" * 70)
print("🎨 1/5 — BASE SD1.5 ANIME CHECKPOINT")
print("=" * 70)

checkpoint = directories["checkpoints"] / "Counterfeit-V3.0_fp16.safetensors"

counterfeit_url = (
    "https://huggingface.co/gsdf/Counterfeit-V3.0/"
    "resolve/main/Counterfeit-V3.0_fp16.safetensors"
)

counterfeit_sha256 = (
    "cbfba64e662370f59d4aa2aa69bf16749fce93846ccce20506aee5df01169859"
)

download_file(
    counterfeit_url,
    checkpoint,
    counterfeit_sha256
)

# ================================================================
# 4. CONTROLNET WEIGHTS
# ================================================================

print("\n" + "=" * 70)
print("🎛️ 2/5 — CONTROLNET")
print("=" * 70)

print("""
ControlNet extension is already installed.

For the final Caesar workflow we want:
  • OpenPose      → character/body posing
  • Lineart Anime → anime structure/line control
  • Depth         → scene geometry/composition

IMPORTANT:
The exact ControlNet filenames/repositories can vary between
SD1.5 model releases and workflow conventions.

Therefore this cell does NOT blindly download guessed ControlNet
weights.

The ControlNet directory is prepared and the workflow can lock
the exact compatible files once the workflow nodes are assembled.
""")

print(f"📁 ControlNet directory ready: {directories['controlnet']}")

# ================================================================
# 5. ANIMATEDIFF
# ================================================================

print("\n" + "=" * 70)
print("🎞️ 3/5 — ANIMATEDIFF")
print("=" * 70)

animatediff = directories["animatediff"] / "mm_sd_v15_v2.ckpt"

if animatediff.exists():
    size = animatediff.stat().st_size / 1024**3
    print(f"✅ AnimateDiff motion model found")
    print(f"   {animatediff}")
    print(f"   {size:.2f} GB")
else:
    print("❌ AnimateDiff motion model missing")
    print(f"   Expected: {animatediff}")

# ================================================================
# 6. IPADAPTER
# ================================================================

print("\n" + "=" * 70)
print("🧑 4/5 — IPADAPTER")
print("=" * 70)

ipadapter_files = [
    directories["ipadapter"] / "ip-adapter-plus_sd15.safetensors",
    directories["ipadapter"] / "ip-adapter-plus-face_sd15.safetensors",
]

clipvision = (
    directories["clip_vision"]
    / "CLIP-ViT-H-14-laion2B-s32B-b79K.safetensors"
)

for f in ipadapter_files:
    if f.exists():
        print(f"✅ {f.name}")
    else:
        print(f"❌ MISSING: {f}")

if clipvision.exists():
    print(f"✅ {clipvision.name}")
else:
    print(f"❌ MISSING: {clipvision}")

# ================================================================
# 7. LIVEPORTRAIT
# ================================================================

print("\n" + "=" * 70)
print("🙂 5/5 — LIVEPORTRAIT")
print("=" * 70)

liveportrait_node = (
    COMFY / "custom_nodes" / "ComfyUI-LivePortraitKJ"
)

if liveportrait_node.exists():
    print("✅ LivePortrait custom node already installed")
    print(f"   {liveportrait_node}")
else:
    print("⚠️ LivePortrait node not found.")
    print("""
Install the KJ LivePortrait node before using LivePortrait.

Its current implementation can automatically download the
compatible safetensor model set into:

    models/liveportrait/
""")

liveportrait_dir = directories["liveportrait"]

print(f"\n📁 LivePortrait model directory:")
print(f"   {liveportrait_dir}")

# ================================================================
# 8. VAE — DO NOT BLINDLY DUPLICATE
# ================================================================

print("\n" + "=" * 70)
print("🧩 VAE")
print("=" * 70)

vae_files = list(directories["vae"].glob("*"))

if vae_files:
    print("Existing VAE files:")
    for f in vae_files:
        if f.is_file():
            print(f"   • {f.name}")
else:
    print("""
ℹ️ No standalone VAE installed.

This is intentional.

We will use the VAE associated with the selected SD1.5
checkpoint/workflow rather than downloading an arbitrary VAE
and potentially creating a compatibility problem.
""")

# ================================================================
# 9. F5-TTS
# ================================================================

print("\n" + "=" * 70)
print("🗣️ F5-TTS")
print("=" * 70)

print("""
F5-TTS remains part of the FINAL production stack because Caesar
needs dialogue.

The actual F5-TTS model will be installed through the F5-TTS
ComfyUI node/model mechanism rather than guessing a model filename
here.

This avoids locking the wrong checkpoint or text encoder before
the dialogue workflow is assembled.
""")

# ================================================================
# 10. TEACACHE
# ================================================================

print("\n" + "=" * 70)
print("⚡ TEACACHE")
print("=" * 70)

print("""
⏭️ NOT INSTALLED.

Reason:
TeaCache is not a universal AnimateDiff/SD1.5 acceleration switch.

Our primary Caesar video pipeline is:
SD1.5 + AnimateDiff.

We will optimize that pipeline through ComfyUI's VRAM/offloading,
context-window and frame settings instead of introducing an
unverified acceleration dependency.
""")

# ================================================================
# 11. FINAL INVENTORY
# ================================================================

print("\n" + "=" * 70)
print("🔍 FINAL MODEL INVENTORY")
print("=" * 70)

checks = {
    "Base SD1.5 checkpoint":
        checkpoint,
    "AnimateDiff motion model":
        animatediff,
    "IPAdapter Plus":
        ipadapter_files[0],
    "IPAdapter Plus Face":
        ipadapter_files[1],
    "CLIP Vision ViT-H":
        clipvision,
}

for name, path in checks.items():
    if path.exists():
        size = path.stat().st_size / 1024**3
        print(f"✅ {name:<30} {size:.2f} GB")
    else:
        print(f"❌ {name:<30} MISSING")

print("\n" + "=" * 70)
print("🎬 CAESAR MODEL LOCK STATUS")
print("=" * 70)

required_now = [
    checkpoint,
    animatediff,
    ipadapter_files[0],
    ipadapter_files[1],
    clipvision,
]

if all(p.exists() for p in required_now):
    print("✅ CORE SD1.5 + ANIMATEDIFF + IPADAPTER STACK READY")
else:
    print("⚠️ CORE STACK IS NOT COMPLETE")

print("""
NEXT PRODUCTION COMPONENTS:

✅ SD1.5 anime base
✅ AnimateDiff
✅ IPAdapter Plus
✅ IPAdapter Face
✅ CLIP Vision
✅ ControlNet infrastructure
✅ Impact Pack infrastructure
✅ VHS infrastructure

🔜 ControlNet workflow-specific weights
🔜 LivePortrait node + weights
🔜 F5-TTS node + model
🔜 Final Caesar workflow JSON

🚫 TeaCache intentionally excluded from the SD1.5 AnimateDiff stack.

DO NOT START GENERATING SCENES YET.
The next step is building the actual Caesar workflow around this
locked core rather than downloading more random weights.
""")

print("=" * 70)
print("🏛️ JULIUS CAESAR MODEL INFRASTRUCTURE CHECK COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# 🎬 JULIUS CAESAR — MASTER COMPONENT INSTALLER
# ControlNet (Fixed URLs) + LivePortrait + F5-TTS
# ================================================================

import os
import sys
import subprocess
import hashlib
from pathlib import Path
from urllib.request import urlopen, Request

COMFY = Path("/content/ComfyUI")
CUSTOM = COMFY / "custom_nodes"
MODELS = COMFY / "models"

CONTROLNET = MODELS / "controlnet"
LIVEPORTRAIT = MODELS / "liveportrait"

CONTROLNET.mkdir(parents=True, exist_ok=True)
LIVEPORTRAIT.mkdir(parents=True, exist_ok=True)
CUSTOM.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("🎬 JULIUS CAESAR — FINAL PRODUCTION COMPONENT INSTALL")
print("=" * 70)


# ================================================================
# HELPERS
# ================================================================

def run(cmd, cwd=None):
    print("\n$", " ".join(cmd))
    result = subprocess.run(cmd, cwd=cwd, text=True)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with return code {result.returncode}")
    return result

def clone_or_update(url, folder):
    folder = Path(folder)
    if folder.exists() and (folder / ".git").exists():
        print(f"⏭️ Already cloned: {folder.name}")
        print("   Pulling latest changes...")
        run(["git", "-C", str(folder), "pull", "--ff-only"])
    else:
        print(f"⬇️ Cloning: {folder.name}")
        run(["git", "clone", "--depth", "1", url, str(folder)])
        print("✅ Clone complete")

def download_verified(filename, url, expected_sha256=None):
    dest = CONTROLNET / filename
    if dest.exists():
        size = dest.stat().st_size / 1024**3
        print(f"⏭️ Already exists: {filename} ({size:.2f} GB)")
        return

    print(f"\n⬇️ Downloading {filename}...")
    tmp = dest.with_suffix(dest.suffix + ".part")
    req = Request(url, headers={"User-Agent": "Mozilla/5.0"})

    with urlopen(req) as response, open(tmp, "wb") as f:
        total = int(response.headers.get("Content-Length", 0))
        done = 0
        while True:
            chunk = response.read(8 * 1024 * 1024)
            if not chunk:
                break
            f.write(chunk)
            done += len(chunk)
            if total:
                print(f"\r   {done/1024**3:.2f} / {total/1024**3:.2f} GB ({done/total*100:.1f}%)", end="")
    print()
    tmp.replace(dest)

    if expected_sha256:
        print("🔐 Checking SHA256...")
        h = hashlib.sha256()
        with open(dest, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)
        actual = h.hexdigest()
        if actual.lower() != expected_sha256.lower():
            dest.unlink(missing_ok=True)
            raise RuntimeError(f"SHA256 mismatch for {filename}\nExpected: {expected_sha256}\nActual:   {actual}")
        print(f"✅ VERIFIED: {filename}")
    else:
        print(f"✅ DOWNLOADED: {filename}")


# ================================================================
# 1. CONTROLNET WEIGHTS (CORRECTED URLS)
# ================================================================

print("\n" + "=" * 70)
print("🎛️ 1/3 — SD1.5 CONTROLNET WEIGHTS")
print("=" * 70)

controlnets = [
    {
        "file": "control_v11p_sd15_openpose.pth",
        "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11p_sd15_openpose.pth",
        "hash": "db97becd92cd19aff71352a60e93c2508decba3dee64f01f686727b9b406a9dd"
    },
    {
        "file": "control_v11f1p_sd15_depth.pth",
        "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11f1p_sd15_depth.pth",
        "hash": None
    },
    {
        "file": "control_v11p_sd15s2_lineart_anime.pth",
        "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11p_sd15s2_lineart_anime.pth",
        "hash": None
    }
]

for cn in controlnets:
    download_verified(cn["file"], cn["url"], cn["hash"])


# ================================================================
# 2. LIVEPORTRAIT
# ================================================================

print("\n" + "=" * 70)
print("🙂 2/3 — LIVEPORTRAIT")
print("=" * 70)

liveportrait_node = CUSTOM / "ComfyUI-LivePortraitKJ"
clone_or_update("https://github.com/kijai/ComfyUI-LivePortraitKJ.git", liveportrait_node)

req_lp = liveportrait_node / "requirements.txt"
if req_lp.exists():
    print("\n📚 Installing LivePortrait requirements...")
    run([sys.executable, "-m", "pip", "install", "-r", str(req_lp)])


# ================================================================
# 3. F5-TTS
# ================================================================

print("\n" + "=" * 70)
print("🗣️ 3/3 — F5-TTS")
print("=" * 70)

f5_node = CUSTOM / "ComfyUI-F5-TTS"
clone_or_update("https://github.com/niknah/ComfyUI-F5-TTS.git", f5_node)

print("\n📦 Initializing F5-TTS submodules...")
run(["git", "submodule", "update", "--init", "--recursive"], cwd=str(f5_node))

req_f5 = f5_node / "requirements.txt"
if req_f5.exists():
    print("\n📚 Installing F5-TTS requirements...")
    run([sys.executable, "-m", "pip", "install", "-r", str(req_f5)])

print("\n" + "=" * 70)
print("✅ ALL COMPONENTS SUCCESSFULLY INSTALLED.")
print("   Restart ComfyUI and refresh the browser.")
print("=" * 70)
```

```python
# ============================================================
# 🟡 LOCALTUNNEL — RUN LAST
# ============================================================

import subprocess
import time
import re

PORT = "8188"

print("=" * 65)
print("🟡 STARTING LOCALTUNNEL")
print("=" * 65)

# Make sure no previous LocalTunnel process is hanging around
subprocess.run(
    "pkill -f 'localtunnel.*8188' || true",
    shell=True
)

time.sleep(2)

print("Connecting to ComfyUI on port 8188...")

tunnel = subprocess.Popen(
    [
        "npx",
        "--yes",
        "localtunnel",
        "--port",
        PORT
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

url = None

start = time.time()

while time.time() - start < 30:
    line = tunnel.stdout.readline()

    if line:
        print(line.strip())

        match = re.search(
            r"https?://[^\s]+",
            line
        )

        if match:
            url = match.group(0)
            break

    time.sleep(0.2)

print("\n" + "=" * 65)

if url:
    print("✅ LOCALTUNNEL READY")
    print("\n🌐 YOUR COMFYUI URL:")
    print(url)
    print("\nOpen that URL in your browser.")
else:
    print("❌ LocalTunnel URL was not detected.")
    print("Tunnel process is still running.")
    print("PID:", tunnel.pid)

print("=" * 65)
```

```python
# =============================================================================
# PROJECT JULIUS CAESAR — IDEMPOTENT COMFYUI INSTALL + MODEL AUDIT CELL
# Designed for an existing ComfyUI install on Colab/Kaggle/Linux.
# Re-run safely: existing repositories/models are preserved; missing items only
# are installed/downloaded. Does NOT start, stop, update, or overwrite ComfyUI.
# =============================================================================

from __future__ import annotations
import os, sys, json, time, shutil, hashlib, subprocess, tempfile, urllib.request
from pathlib import Path
from datetime import datetime, timezone

# ------------------------------- USER SETTINGS -------------------------------
COMFYUI_DIR_OVERRIDE = None  # Example: "/content/dagshub_storage/comfyui_project/ComfyUI"
INSTALL_CUSTOM_NODES = True
INSTALL_NODE_REQUIREMENTS = True
INSTALL_FFMPEG = True
DOWNLOAD_MODEL_ASSETS = True       # Large download (~15 GB when all are missing)
MIN_FREE_SPACE_GB = 3.0            # Hard stop if free space is below this amount
VERIFY_SHA256_ON_DOWNLOAD = True
# Safety: this cell will NOT git-pull/update existing repositories, upgrade ComfyUI,
# or intentionally upgrade the currently installed torch/torchvision/torchaudio/numpy.

# --------------------------------- HELPERS -----------------------------------
def say(msg):
    print(f"[JC setup] {msg}", flush=True)

def run_cmd(cmd, *, cwd=None, check=False, capture=False):
    say("$ " + " ".join(str(x) for x in cmd))
    result = subprocess.run(
        [str(x) for x in cmd], cwd=str(cwd) if cwd else None,
        text=True, stdout=subprocess.PIPE if capture else None,
        stderr=subprocess.STDOUT if capture else None,
    )
    if capture and result.stdout:
        print(result.stdout[-12000:])
    if check and result.returncode != 0:
        raise RuntimeError(f"Command failed ({result.returncode}): {' '.join(map(str, cmd))}")
    return result

def find_comfy_root():
    if COMFYUI_DIR_OVERRIDE:
        p = Path(COMFYUI_DIR_OVERRIDE).expanduser().resolve()
        if not (p / "main.py").exists() or not (p / "custom_nodes").exists():
            raise FileNotFoundError(f"COMFYUI_DIR_OVERRIDE is not an existing ComfyUI root: {p}")
        return p

    env_path = os.environ.get("COMFYUI_DIR")
    candidates = [
        Path(env_path) if env_path else None,
        Path("/content/dagshub_storage/comfyui_project/ComfyUI"),
        Path("/content/ComfyUI"),
        Path("/kaggle/working/comfyui_project/ComfyUI"),
        Path("/kaggle/working/ComfyUI"),
        Path("/workspace/ComfyUI"),
        Path.cwd() / "ComfyUI",
        Path.cwd(),
    ]
    for p in candidates:
        if p and p.is_dir() and (p / "main.py").is_file() and (p / "custom_nodes").is_dir():
            return p.resolve()
    raise FileNotFoundError(
        "Could not find an existing ComfyUI root. Set COMFYUI_DIR_OVERRIDE at the top of this cell. "
        "The installer deliberately will not create a second ComfyUI installation."
    )

def sha256_file(path: Path, chunk_size=8 * 1024 * 1024):
    h = hashlib.sha256()
    with path.open("rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)
    return h.hexdigest()

def install_repo(repo_name, url, recursive=False):
    dest = custom_nodes / repo_name
    if dest.exists():
        if (dest / ".git").exists():
            origin = run_cmd(["git", "remote", "get-url", "origin"], cwd=dest, capture=True)
            origin_text = (origin.stdout or "").strip().splitlines()[-1] if origin.stdout else "unknown origin"
            say(f"EXISTS — preserving {repo_name}; origin={origin_text}. No pull/update performed.")
            return dest, True
        # Do not overwrite folders that may contain user changes or a Manager-managed install.
        say(f"WARNING — {dest} exists but is not a Git checkout. Preserving it; cannot verify/update origin.")
        return dest, False
    cmd = ["git", "clone", "--depth", "1"]
    if recursive:
        cmd = ["git", "clone", "--depth", "1", "--recurse-submodules"]
    cmd += [url, str(dest)]
    result = run_cmd(cmd, check=False, capture=True)
    if result.returncode != 0:
        say(f"ERROR — clone failed for {repo_name}; see output above.")
        return dest, False
    if recursive:
        sub = run_cmd(["git", "submodule", "update", "--init", "--recursive"], cwd=dest, check=False, capture=True)
        if sub.returncode != 0:
            say(f"WARNING — recursive submodule init failed for {repo_name}.")
    say(f"CLONED — {repo_name}")
    return dest, True

def make_constraints_file():
    """Pin existing core runtime packages so node requirements cannot replace them."""
    pins = []
    try:
        from importlib.metadata import version, PackageNotFoundError
        protected = ["torch", "torchvision", "torchaudio", "numpy"]
        for pkg in protected:
            try:
                v = version(pkg)
                # Exact constraints also protect local build suffixes such as +cu121.
                pins.append(f"{pkg}=={v}")
            except PackageNotFoundError:
                pass
    except Exception as e:
        say(f"Could not collect runtime constraints: {e}")
    fd, path = tempfile.mkstemp(prefix="jc_pip_constraints_", suffix=".txt")
    with os.fdopen(fd, "w", encoding="utf-8") as f:
        f.write("\n".join(pins) + ("\n" if pins else ""))
    say("Protected runtime packages: " + (", ".join(pins) if pins else "none detected"))
    return Path(path)

def requirements_fingerprint(req_path: Path):
    return hashlib.sha256(req_path.read_bytes()).hexdigest()

def install_requirements(repo_name, repo_path, constraints, state):
    req = repo_path / "requirements.txt"
    if not req.is_file():
        say(f"NO requirements.txt — {repo_name}; nothing to install here.")
        return True
    try:
        req_hash = requirements_fingerprint(req)
    except Exception:
        req_hash = "unreadable"
    try:
        import torch
        torch_v = getattr(torch, "__version__", "not-installed")
    except Exception:
        torch_v = "not-installed"
    key = f"{repo_name}|{req_hash}|py{sys.version_info.major}.{sys.version_info.minor}|torch{torch_v}"
    if state.get("requirements_ok", {}).get(key) is True:
        say(f"REQUIREMENTS ALREADY INSTALLED — {repo_name} (same requirements + runtime fingerprint).")
        return True
    say(f"Installing requirements for {repo_name} with constraints protecting current Torch/NumPy...")
    result = run_cmd(
        [sys.executable, "-m", "pip", "install", "--disable-pip-version-check",
         "--upgrade-strategy", "only-if-needed", "-c", str(constraints), "-r", str(req)],
        check=False, capture=True,
    )
    if result.returncode == 0:
        state.setdefault("requirements_ok", {})[key] = True
        say(f"REQUIREMENTS OK — {repo_name}")
        return True
    say(f"WARNING — requirements install failed for {repo_name}. Existing runtime was protected; inspect the error above.")
    state.setdefault("requirements_failed", {})[key] = datetime.now(timezone.utc).isoformat()
    return False

def resumable_download(url: str, target: Path, min_bytes: int, *, expected_sha256=None, label=None):
    label = label or target.name
    target.parent.mkdir(parents=True, exist_ok=True)
    if target.is_file() and target.stat().st_size >= min_bytes:
        if VERIFY_SHA256_ON_DOWNLOAD and expected_sha256:
            say(f"Verifying existing file SHA-256 for {label}...")
            actual_sha = sha256_file(target)
            if actual_sha.lower() == expected_sha256.lower():
                say(f"EXISTS + SHA-256 OK — {label}; leaving unchanged.")
                return True
            bad = target.with_name(target.name + f".sha256-mismatch-{int(time.time())}")
            target.rename(bad)
            say(f"WARNING — existing {label} failed SHA-256; preserved as {bad.name}; downloading a fresh copy.")
        else:
            say(f"EXISTS — {label} ({target.stat().st_size / (1024**3):.2f} GiB); leaving unchanged.")
            return True
    if target.exists() and (not target.is_file() or target.stat().st_size < min_bytes):
        bad = target.with_name(target.name + f".invalid-{int(time.time())}")
        say(f"WARNING — existing target looks incomplete; preserving it as {bad.name}")
        target.rename(bad)

    part = target.with_name(target.name + ".part")
    existing = part.stat().st_size if part.exists() else 0
    headers = {"User-Agent": "ProjectJuliusCaesar-IdempotentInstaller/1.0"}
    if existing:
        headers["Range"] = f"bytes={existing}-"
        say(f"RESUMING — {label} from {existing / (1024**2):.1f} MiB")
    else:
        say(f"DOWNLOADING — {label}")
    request = urllib.request.Request(url, headers=headers)
    try:
        with urllib.request.urlopen(request, timeout=60) as response:
            status = getattr(response, "status", response.getcode())
            # Append only if server honored the byte-range request.
            mode = "ab" if existing and status == 206 else "wb"
            if mode == "wb" and existing:
                say(f"Server did not accept resume for {label}; restarting this partial file.")
            written = 0
            last_report = 0
            with part.open(mode) as out:
                while True:
                    block = response.read(8 * 1024 * 1024)
                    if not block:
                        break
                    out.write(block)
                    written += len(block)
                    if written - last_report >= 256 * 1024 * 1024:
                        total = part.stat().st_size / (1024**3)
                        say(f"  {label}: downloaded/available {total:.2f} GiB")
                        last_report = written
    except Exception as e:
        say(f"DOWNLOAD INTERRUPTED — {label}: {e}")
        say("The .part file is retained so a later rerun can attempt to resume it.")
        return False

    actual_size = part.stat().st_size if part.exists() else 0
    if actual_size < min_bytes:
        say(f"ERROR — {label} is only {actual_size} bytes (expected at least {min_bytes}); leaving it as .part.")
        return False
    if VERIFY_SHA256_ON_DOWNLOAD and expected_sha256:
        say(f"Verifying SHA-256 for {label} (this reads the whole file once)...")
        actual_sha = sha256_file(part)
        if actual_sha.lower() != expected_sha256.lower():
            bad = part.with_name(part.name + ".sha256-mismatch")
            part.rename(bad)
            say(f"ERROR — SHA-256 mismatch for {label}; preserved as {bad.name}; will not install it.")
            return False
    os.replace(part, target)
    say(f"SAVED — {label} ({target.stat().st_size / (1024**3):.2f} GiB)")
    return True

# ------------------------------- PRE-FLIGHT ---------------------------------
try:
    import torch
    torch_report = f"torch={torch.__version__}, cuda_available={torch.cuda.is_available()}"
    if torch.cuda.is_available():
        torch_report += f", gpu={torch.cuda.get_device_name(0)}"
except Exception as e:
    torch_report = f"Torch import failed: {e}"

root = find_comfy_root()
custom_nodes = root / "custom_nodes"
models = root / "models"
custom_nodes.mkdir(parents=True, exist_ok=True)
models.mkdir(parents=True, exist_ok=True)
state_path = root / ".project_julius_installer_state.json"
try:
    state = json.loads(state_path.read_text(encoding="utf-8")) if state_path.exists() else {}
except Exception:
    state = {}
state.setdefault("requirements_ok", {})
state.setdefault("requirements_failed", {})
state.setdefault("extension_status", {})
state.setdefault("model_status", {})

say(f"ComfyUI root: {root}")
say(f"Python: {sys.version.split()[0]} | {torch_report}")
say("This cell does not start/stop ComfyUI and never pulls updates into existing Git repos.")

# ------------------------------- SYSTEM TOOLS -------------------------------
if INSTALL_FFMPEG:
    if shutil.which("ffmpeg"):
        say(f"FFmpeg already available: {shutil.which('ffmpeg')}")
    elif shutil.which("apt-get"):
        say("FFmpeg is missing; attempting an apt install without upgrading other packages.")
        update = run_cmd(["apt-get", "update", "-qq"], check=False, capture=True)
        install = run_cmd(["apt-get", "install", "-y", "-qq", "ffmpeg"], check=False, capture=True)
        if install.returncode == 0 and shutil.which("ffmpeg"):
            say("FFmpeg installed successfully.")
        else:
            say("WARNING — could not install FFmpeg automatically; install it at the OS level before video assembly.")
    else:
        say("WARNING — ffmpeg missing and apt-get unavailable; install it for video/audio processing.")

# ------------------------------- NODE REPOS ---------------------------------
REPOS = [
    ("ComfyUI-AnimateDiff-Evolved", "https://github.com/Kosinkadink/ComfyUI-AnimateDiff-Evolved.git", False),
    ("ComfyUI_IPAdapter_plus", "https://github.com/cubiq/ComfyUI_IPAdapter_plus.git", False),
    ("comfyui_controlnet_aux", "https://github.com/Fannovel16/comfyui_controlnet_aux.git", False),
    ("ComfyUI-Impact-Pack", "https://github.com/ltdrdata/ComfyUI-Impact-Pack.git", False),
    ("ComfyUI-Impact-Subpack", "https://github.com/ltdrdata/ComfyUI-Impact-Subpack.git", False),
    ("comfyui-character-composer", "https://github.com/unh1nge/comfyui-character-composer.git", False),
    ("ComfyUI-F5-TTS", "https://github.com/niknah/ComfyUI-F5-TTS.git", True),
    ("ComfyUI-MMAudio", "https://github.com/kijai/ComfyUI-MMAudio.git", False),
    ("ComfyUI_wav2lip", "https://github.com/ShmuelRonen/ComfyUI_wav2lip.git", False),
    ("ComfyUI-VideoHelperSuite", "https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite.git", False),
    ("ComfyUI-TeaCache", "https://github.com/welltop-cn/ComfyUI-TeaCache.git", False),
]
repo_paths = {}
if INSTALL_CUSTOM_NODES:
    if not shutil.which("git"):
        raise RuntimeError("git is missing; install git before running this cell.")
    for name, url, recursive in REPOS:
        repo_path, okay = install_repo(name, url, recursive=recursive)
        repo_paths[name] = repo_path
        state["extension_status"][name] = "present" if okay else "needs_attention"

# F5-TTS submodule fix: safe to rerun, does not wipe existing files.
f5_path = custom_nodes / "ComfyUI-F5-TTS"
if f5_path.is_dir() and (f5_path / ".git").exists():
    say("Ensuring F5-TTS submodules are initialized...")
    sub = run_cmd(["git", "submodule", "update", "--init", "--recursive"], cwd=f5_path, check=False, capture=True)
    if sub.returncode != 0:
        say("WARNING — F5-TTS submodule update failed. F5-TTS may need manual repair; existing files were not deleted.")

# Requirements install is per-extension; constraints protect installed torch/vision/audio/numpy.
constraints_path = make_constraints_file()
if INSTALL_NODE_REQUIREMENTS:
    order = [name for name, _, _ in REPOS]
    for name in order:
        p = custom_nodes / name
        if p.is_dir():
            install_requirements(name, p, constraints_path, state)

# ------------------------------- MODEL MANIFEST -----------------------------
# URL, relative destination, minimum valid byte size, expected SHA-256, size estimate GiB.
# Exact SHA checks are included for the official ViT-H encoder and MMAudio main weight.
MODEL_ASSETS = [
    {
        "key": "sd15_checkpoint", "url": "https://huggingface.co/gsdf/Counterfeit-V3.0/resolve/main/Counterfeit-V3.0_fp16.safetensors?download=true",
        "path": "models/checkpoints/Counterfeit-V3.0_fp16.safetensors", "min": 1_000_000_000, "sha256": None, "estimate_gib": 2.1,
    },
    {
        "key": "vae", "url": "https://huggingface.co/stabilityai/sd-vae-ft-mse-original/resolve/main/vae-ft-mse-840000-ema-pruned.safetensors?download=true",
        "path": "models/vae/vae-ft-mse-840000-ema-pruned.safetensors", "min": 100_000_000, "sha256": None, "estimate_gib": 0.32,
    },
    {
        "key": "animatediff_motion", "url": "https://huggingface.co/guoyww/animatediff/resolve/main/mm_sd_v15_v2.ckpt?download=true",
        "path": "models/animatediff_models/mm_sd_v15_v2.ckpt", "min": 500_000_000, "sha256": None, "estimate_gib": 1.7,
    },
    {
        "key": "openpose_controlnet", "url": "https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11p_sd15_openpose.pth?download=true",
        "path": "models/controlnet/control_v11p_sd15_openpose.pth", "min": 500_000_000, "sha256": None, "estimate_gib": 1.4,
    },
    {
        "key": "ipadapter_plus", "url": "https://huggingface.co/h94/IP-Adapter/resolve/main/models/ip-adapter-plus_sd15.safetensors?download=true",
        "path": "models/ipadapter/ip-adapter-plus_sd15.safetensors", "min": 40_000_000, "sha256": "a1c250be40455cc61a43da1201ec3f1edaea71214865fb47f57927e06cbe4996", "estimate_gib": 0.15,
    },
    {
        "key": "ipadapter_plus_face", "url": "https://huggingface.co/h94/IP-Adapter/resolve/main/models/ip-adapter-plus-face_sd15.safetensors?download=true",
        "path": "models/ipadapter/ip-adapter-plus-face_sd15.safetensors", "min": 40_000_000, "sha256": "1c9edc21af6f737dc1d6e0e734190e976cfacf802d6b024b77aa3be922f7569b", "estimate_gib": 0.10,
    },
    {
        "key": "clip_vision_vit_h", "url": "https://huggingface.co/h94/IP-Adapter/resolve/main/models/image_encoder/model.safetensors?download=true",
        "path": "models/clip_vision/CLIP-ViT-H-14-laion2B-s32B-b79K.safetensors", "min": 1_000_000_000, "sha256": "6ca9667da1ca9e0b0f75e46bb030f7e011f44f86cbfb8d5a36590fcd7507b030", "estimate_gib": 2.36,
    },
    {
        "key": "f5_tts_model", "url": "https://huggingface.co/SWivid/F5-TTS/resolve/main/F5TTS_v1_Base/model_1250000.safetensors?download=true",
        "path": "models/checkpoints/F5-TTS/model_1250000.safetensors", "min": 500_000_000, "sha256": None, "estimate_gib": 1.3,
    },
    {
        "key": "f5_tts_vocab", "url": "https://huggingface.co/SWivid/F5-TTS/resolve/main/F5TTS_v1_Base/vocab.txt?download=true",
        "path": "models/checkpoints/F5-TTS/vocab.txt", "min": 10, "sha256": None, "estimate_gib": 0.001,
    },
    {
        "key": "mmaudio_large_44k", "url": "https://huggingface.co/Kijai/MMAudio_safetensors/resolve/main/mmaudio_large_44k_v2_fp16.safetensors?download=true",
        "path": "models/mmaudio/mmaudio_large_44k_v2_fp16.safetensors", "min": 1_000_000_000, "sha256": "a61378951ba4e8bfde2e977c0ad8211a7e02e1e92fd6a5b1c258a6464c1d102f", "estimate_gib": 1.92,
    },
    {
        "key": "mmaudio_synchformer", "url": "https://huggingface.co/Kijai/MMAudio_safetensors/resolve/main/mmaudio_synchformer_fp16.safetensors?download=true",
        "path": "models/mmaudio/mmaudio_synchformer_fp16.safetensors", "min": 100_000_000, "sha256": None, "estimate_gib": 0.45,
    },
    {
        "key": "mmaudio_vae", "url": "https://huggingface.co/Kijai/MMAudio_safetensors/resolve/main/mmaudio_vae_44k_fp16.safetensors?download=true",
        "path": "models/mmaudio/mmaudio_vae_44k_fp16.safetensors", "min": 100_000_000, "sha256": None, "estimate_gib": 0.58,
    },
    {
        "key": "mmaudio_clip", "url": "https://huggingface.co/Kijai/MMAudio_safetensors/resolve/main/apple_DFN5B-CLIP-ViT-H-14-384_fp16.safetensors?download=true",
        "path": "models/mmaudio/apple_DFN5B-CLIP-ViT-H-14-384_fp16.safetensors", "min": 1_000_000_000, "sha256": None, "estimate_gib": 1.84,
    },
    {
        "key": "wav2lip_checkpoint", "url": "https://huggingface.co/Nekochu/Wav2Lip/resolve/main/wav2lip_gan.pth?download=true",
        "path": "custom_nodes/ComfyUI_wav2lip/Wav2Lip/checkpoints/wav2lip_gan.pth", "min": 300_000_000,
        "sha256": "ca9ab7b7b812c0e80a6e70a5977c545a1e8a365a6c49d5e533023c034d7ac3d8", "estimate_gib": 0.41,
    },
]

# Required folders even where the extension auto-downloads its own model assets.
for rel in [
    "models/checkpoints", "models/vae", "models/animatediff_models", "models/controlnet",
    "models/ipadapter", "models/clip_vision", "models/mmaudio", "models/ultralytics/bbox",
    "models/mmaudio/nvidia/bigvgan_v2_44khz_128band_512x",
]:
    (root / rel).mkdir(parents=True, exist_ok=True)

if DOWNLOAD_MODEL_ASSETS:
    missing = []
    estimated_missing_gib = 0.0
    for item in MODEL_ASSETS:
        target = root / item["path"]
        if target.is_file() and target.stat().st_size >= item["min"]:
            continue
        missing.append(item)
        estimated_missing_gib += float(item["estimate_gib"])
    free_gib = shutil.disk_usage(root).free / (1024**3)
    say(f"Model download preflight: {len(missing)} missing assets; estimated {estimated_missing_gib:.1f} GiB; free disk {free_gib:.1f} GiB.")
    if free_gib < MIN_FREE_SPACE_GB:
        raise RuntimeError(f"Only {free_gib:.1f} GiB free; safety threshold is {MIN_FREE_SPACE_GB:.1f} GiB. Free space or disable DOWNLOAD_MODEL_ASSETS.")
    if missing and free_gib < estimated_missing_gib + 1.0:
        raise RuntimeError(
            f"Not enough free space for estimated model downloads: need about {estimated_missing_gib + 1.0:.1f} GiB incl. buffer, "
            f"have {free_gib:.1f} GiB. No model downloads were started. Change model paths/storage or free space."
        )
    for item in MODEL_ASSETS:
        target = root / item["path"]
        ok = resumable_download(
            item["url"], target, item["min"], expected_sha256=item["sha256"], label=item["key"]
        )
        state["model_status"][item["key"]] = "ready" if ok else "needs_attention"
else:
    say("DOWNLOAD_MODEL_ASSETS=False: model downloads skipped; audit will still list missing assets.")

# Wav2Lip checkpoint is now part of MODEL_ASSETS and uses the source linked from its node README.
wav2lip_ckpt = custom_nodes / "ComfyUI_wav2lip" / "Wav2Lip" / "checkpoints" / "wav2lip_gan.pth"
wav2lip_ckpt.parent.mkdir(parents=True, exist_ok=True)
if wav2lip_ckpt.is_file() and wav2lip_ckpt.stat().st_size > 300_000_000:
    state["model_status"]["wav2lip_checkpoint"] = "ready"
else:
    state["model_status"]["wav2lip_checkpoint"] = "needs_attention"
    say(f"ACTION REQUIRED — verify Wav2Lip checkpoint download: {wav2lip_ckpt}")

# State/report writes are atomic and safe to repeat.
state["last_run_utc"] = datetime.now(timezone.utc).isoformat()
state["comfyui_root"] = str(root)
state["python_version"] = sys.version.split()[0]
state["runtime"] = torch_report
state["teacache"] = "installed if repo clone succeeded; node availability requires ComfyUI restart"
state["important_note"] = "Restart ComfyUI before API/node-schema checks. This cell has not validated workflow compatibility or generation quality."
tmp_state = state_path.with_suffix(".tmp")
tmp_state.write_text(json.dumps(state, indent=2), encoding="utf-8")
os.replace(tmp_state, state_path)

# ------------------------------- FINAL AUDIT --------------------------------
say("=" * 76)
say("FINAL FILE AUDIT")
missing_assets = []
for item in MODEL_ASSETS:
    p = root / item["path"]
    ready = p.is_file() and p.stat().st_size >= item["min"]
    say(("OK      " if ready else "MISSING ") + f"{item['key']}: {p}")
    if not ready:
        missing_assets.append(item["key"])
face_detector = root / "models/ultralytics/bbox/face_yolov8m.pt"
say(("OK      " if face_detector.is_file() else "AUTO/WAIT ") + f"face detector: {face_detector}")
say(("OK      " if shutil.which("ffmpeg") else "MISSING ") + "ffmpeg")
say(("OK      " if wav2lip_ckpt.is_file() and wav2lip_ckpt.stat().st_size > 100_000_000 else "ACTION  ") + f"Wav2Lip checkpoint: {wav2lip_ckpt}")

repo_missing = []
for name, url, _ in REPOS:
    p = custom_nodes / name
    present = p.is_dir() and (p / "__init__.py").exists()
    say(("PRESENT " if present else "CHECK   ") + f"node folder: {name}")
    if not p.is_dir():
        repo_missing.append(name)

report_path = root / "project_julius_installer_report.json"
report = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "comfyui_root": str(root),
    "python": sys.version.split()[0],
    "torch_runtime": torch_report,
    "download_models_enabled": DOWNLOAD_MODEL_ASSETS,
    "missing_model_assets": missing_assets,
    "missing_repository_folders": repo_missing,
    "requirements_failures": state.get("requirements_failed", {}),
    "requirements_installation_enabled": INSTALL_NODE_REQUIREMENTS,
    "wav2lip_checkpoint": str(wav2lip_ckpt),
    "face_detector_path": str(face_detector),
    "ffmpeg_path": shutil.which("ffmpeg"),
    "next_step": "Restart ComfyUI, then check /object_info and run a small workflow test. Node imports and end-to-end compatibility are not proven by file presence alone.",
}
report_path.write_text(json.dumps(report, indent=2), encoding="utf-8")
say(f"Installer state: {state_path}")
say(f"Audit report:   {report_path}")
say("DONE — restart ComfyUI before checking newly installed nodes. No existing Git repo was pulled or overwritten.")
```

```python
#!/usr/bin/env python3
"""Julius Caesar / ComfyUI scene-and-shot orchestrator.

Safety defaults:
- This script NEVER starts, stops, or restarts ComfyUI.
- Without --execute, it is a dry run: no ComfyUI jobs and no Dagshub writes.
- The script does not install Python packages or download models.
- A real run requires a ComfyUI API-format workflow JSON and explicit node/field bindings.

Locally syntax-checked and passed CPU-only unit tests with a mocked ComfyUI API server.
Not tested against a live ComfyUI server, Kaggle GPU, or the user's DagsHub bucket.
"""
from __future__ import annotations

import argparse
import copy
import hashlib
import json
import mimetypes
import os
import re
import shutil
import sys
import time
import urllib.error
import urllib.request
from concurrent.futures import ThreadPoolExecutor, as_completed
from dataclasses import dataclass, asdict
from datetime import datetime, timezone
from pathlib import Path, PurePosixPath
from typing import Any, Dict, Iterable, List, Optional, Sequence, Tuple

VERSION = "0.1.0"
SHOT_HEADER_RE = re.compile(r"^\s*(?:shot\s*)#?0*(\d+)\s*[:.\-]\s*(.*)$", re.I)
VALID_SOURCES = {
    "prompt", "shot_prompt", "negative_prompt", "dialogue", "characters", "camera",
    "continuity", "scene_id", "scene_name", "shot_id", "shot_number", "shot_index",
    "shot_count", "seed", "duration_seconds", "filename_prefix",
}


class OrchestratorError(RuntimeError):
    """Expected, user-actionable orchestrator error."""


def now_utc() -> str:
    return datetime.now(timezone.utc).isoformat(timespec="seconds")


def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()


def sha256_file(path: Path, chunk_size: int = 1024 * 1024) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            digest.update(chunk)
    return digest.hexdigest()


def canonical_json_bytes(value: Any) -> bytes:
    return json.dumps(value, sort_keys=True, ensure_ascii=False, separators=(",", ":")).encode("utf-8")


def safe_slug(value: str, fallback: str = "item") -> str:
    value = value.strip().lower()
    value = re.sub(r"[^a-z0-9._-]+", "_", value)
    value = value.strip("._-")
    return (value or fallback)[:100]


def load_json(path: Path) -> Any:
    try:
        with path.open("r", encoding="utf-8") as f:
            return json.load(f)
    except FileNotFoundError:
        raise OrchestratorError(f"File not found: {path}")
    except json.JSONDecodeError as e:
        raise OrchestratorError(f"Invalid JSON in {path} (line {e.lineno}, column {e.colno}): {e.msg}")


def resolve_path(raw: str, base_dir: Path) -> Path:
    path = Path(raw).expanduser()
    return path if path.is_absolute() else (base_dir / path).resolve()


@dataclass
class Shot:
    shot_id: str
    description: str
    dialogue: str = ""
    characters: str = ""
    camera: str = ""
    continuity: str = ""
    negative_prompt: str = ""
    duration_seconds: float = 0.0

    @classmethod
    def from_obj(cls, obj: Dict[str, Any], index: int) -> "Shot":
        if not isinstance(obj, dict):
            raise OrchestratorError(f"Shot {index} must be a JSON object.")
        description = str(obj.get("description", obj.get("prompt", ""))).strip()
        if not description:
            raise OrchestratorError(f"Shot {index} has no description/prompt.")
        shot_id = safe_slug(str(obj.get("shot_id", f"shot_{index:03d}")), f"shot_{index:03d}")
        try:
            duration = float(obj.get("duration_seconds", 0.0) or 0.0)
        except (TypeError, ValueError):
            raise OrchestratorError(f"Shot {shot_id}: duration_seconds must be numeric.")
        if duration < 0:
            raise OrchestratorError(f"Shot {shot_id}: duration_seconds cannot be negative.")
        return cls(
            shot_id=shot_id,
            description=description,
            dialogue=str(obj.get("dialogue", "")).strip(),
            characters=str(obj.get("characters", "")).strip(),
            camera=str(obj.get("camera", "")).strip(),
            continuity=str(obj.get("continuity", "")).strip(),
            negative_prompt=str(obj.get("negative_prompt", "")).strip(),
            duration_seconds=duration,
        )


def parse_shot_text(text: str, expected_count: int) -> List[Shot]:
    """Parse SHOT 001: ... sections, or one non-empty line per shot."""
    lines = text.splitlines()
    sections: List[Tuple[int, List[str]]] = []
    current_num: Optional[int] = None
    current_lines: List[str] = []
    has_headers = any(SHOT_HEADER_RE.match(line) for line in lines)
    if has_headers:
        for line in lines:
            match = SHOT_HEADER_RE.match(line)
            if match:
                if current_num is not None:
                    sections.append((current_num, current_lines))
                current_num = int(match.group(1))
                current_lines = [match.group(2).strip()] if match.group(2).strip() else []
            elif current_num is not None:
                if line.strip():
                    current_lines.append(line.strip())
            elif line.strip():
                raise OrchestratorError(
                    "Text before the first SHOT heading. Use 'SHOT 001: description' headings, "
                    "or use one non-empty line per shot."
                )
        if current_num is not None:
            sections.append((current_num, current_lines))
        raw_items = ["\n".join(body).strip() for _, body in sections]
    else:
        raw_items = [line.strip() for line in lines if line.strip()]

    raw_items = [x for x in raw_items if x]
    if len(raw_items) != expected_count:
        raise OrchestratorError(
            f"You specified {expected_count} shots, but the pasted text parsed into {len(raw_items)} shots. "
            "Use one non-empty line per shot, or label multi-line shots as 'SHOT 001: ...', and try again."
        )
    return [Shot.from_obj({"shot_id": f"shot_{i:03d}", "description": desc}, i)
            for i, desc in enumerate(raw_items, start=1)]


def read_shot_file(path: Path) -> List[Shot]:
    if path.suffix.lower() == ".json":
        raw = load_json(path)
        if isinstance(raw, dict):
            raw = raw.get("shots")
        if not isinstance(raw, list):
            raise OrchestratorError("Shot JSON must be a list, or an object with a 'shots' list.")
        shots = [Shot.from_obj(item, i) for i, item in enumerate(raw, start=1)]
    else:
        text = path.read_text(encoding="utf-8-sig")
        lines = [line for line in text.splitlines() if line.strip()]
        if not lines:
            raise OrchestratorError(f"Shot file is empty: {path}")
        shots = parse_shot_text(text, len(lines) if not any(SHOT_HEADER_RE.match(x) for x in text.splitlines()) else sum(bool(SHOT_HEADER_RE.match(x)) for x in text.splitlines()))
    if not shots:
        raise OrchestratorError(f"No shots found in {path}")
    ensure_unique_shot_ids(shots)
    return shots


def ensure_unique_shot_ids(shots: Sequence[Shot]) -> None:
    ids = [s.shot_id for s in shots]
    if len(ids) != len(set(ids)):
        raise OrchestratorError("Shot IDs must be unique within a scene. Check the shot JSON file.")


def ask(prompt: str, default: Optional[str] = None) -> str:
    suffix = f" [{default}]" if default is not None else ""
    value = input(f"{prompt}{suffix}: ").strip()
    return value if value else (default or "")


def ask_int(prompt: str, minimum: int = 1, maximum: Optional[int] = None, default: Optional[int] = None) -> int:
    while True:
        raw = ask(prompt, str(default) if default is not None else None)
        try:
            val = int(raw)
            if val < minimum or (maximum is not None and val > maximum):
                bound = f"{minimum}–{maximum}" if maximum is not None else f"at least {minimum}"
                print(f"Enter a number in the range {bound}.")
                continue
            return val
        except ValueError:
            print("Enter a whole number.")


def gather_shots_interactive(count: int) -> List[Shot]:
    print("\nPaste the shot-by-shot breakdown below.")
    print("Format A: one shot per non-empty line.")
    print("Format B: multi-line shots headed by 'SHOT 001: description'.")
    print("Type END on a line by itself when finished. Nothing is rendered until you choose --execute.\n")
    lines: List[str] = []
    while True:
        line = input()
        if line.strip().upper() == "END":
            break
        lines.append(line)
    return parse_shot_text("\n".join(lines), count)


class ArtifactStore:
    """Minimal interface implemented by local and DagsHub S3 storage backends."""
    def get_bytes(self, key: str) -> Optional[bytes]:
        raise NotImplementedError

    def put_bytes(self, key: str, data: bytes, content_type: str = "application/octet-stream") -> None:
        raise NotImplementedError

    def upload_file(self, local_path: Path, key: str, sha256: str) -> Dict[str, Any]:
        raise NotImplementedError

    def verify_object(self, record: Dict[str, Any]) -> bool:
        raise NotImplementedError

    def close(self) -> None:
        pass


class LocalArtifactStore(ArtifactStore):
    def __init__(self, root: Path):
        self.root = root.resolve()
        self.root.mkdir(parents=True, exist_ok=True)

    def _path(self, key: str) -> Path:
        clean = PurePosixPath(key)
        if clean.is_absolute() or ".." in clean.parts:
            raise OrchestratorError(f"Unsafe storage key: {key}")
        path = (self.root / Path(*clean.parts)).resolve()
        if self.root not in path.parents and path != self.root:
            raise OrchestratorError(f"Storage key escaped local root: {key}")
        return path

    def get_bytes(self, key: str) -> Optional[bytes]:
        path = self._path(key)
        return path.read_bytes() if path.is_file() else None

    def put_bytes(self, key: str, data: bytes, content_type: str = "application/octet-stream") -> None:
        path = self._path(key)
        path.parent.mkdir(parents=True, exist_ok=True)
        temp = path.with_name(path.name + ".uploading")
        temp.write_bytes(data)
        os.replace(temp, path)

    def upload_file(self, local_path: Path, key: str, sha256: str) -> Dict[str, Any]:
        path = self._path(key)
        path.parent.mkdir(parents=True, exist_ok=True)
        temp = path.with_name(path.name + ".uploading")
        shutil.copy2(local_path, temp)
        os.replace(temp, path)
        return {"key": key, "size": path.stat().st_size, "sha256": sha256}

    def verify_object(self, record: Dict[str, Any]) -> bool:
        try:
            path = self._path(str(record["key"]))
            if not path.is_file() or path.stat().st_size != int(record["size"]):
                return False
            return sha256_file(path) == record["sha256"]
        except (KeyError, ValueError, OSError, OrchestratorError):
            return False


class DagshubBucketStore(ArtifactStore):
    """DagsHub's S3-compatible repository bucket (not Git/DVC repository files)."""
    def __init__(self, repo: str):
        if "/" not in repo:
            raise OrchestratorError("DagsHub repo must be 'owner/repository'.")
        try:
            from dagshub import get_repo_bucket_client
        except ImportError as e:
            raise OrchestratorError(
                "DagsHub bucket mode needs the 'dagshub' package. Install it in the existing runtime "
                "before executing; this script deliberately does not install dependencies."
            ) from e
        self.repo = repo
        self.bucket = repo.split("/", 1)[1]
        try:
            self.client = get_repo_bucket_client(repo, flavor="boto")
        except Exception as e:
            raise OrchestratorError(
                f"Could not create the DagsHub bucket client for {repo}: {e}. "
                "Authenticate using the notebook's established DagsHub authentication step and retry."
            ) from e

    @staticmethod
    def _not_found(exc: Exception) -> bool:
        response = getattr(exc, "response", {})
        code = str(response.get("Error", {}).get("Code", ""))
        status = response.get("ResponseMetadata", {}).get("HTTPStatusCode")
        return code in {"404", "NoSuchKey", "NotFound"} or status == 404

    def get_bytes(self, key: str) -> Optional[bytes]:
        try:
            response = self.client.get_object(Bucket=self.bucket, Key=key)
            return response["Body"].read()
        except Exception as e:
            if self._not_found(e):
                return None
            raise OrchestratorError(f"DagsHub GET failed for '{key}': {e}") from e

    def put_bytes(self, key: str, data: bytes, content_type: str = "application/octet-stream") -> None:
        digest = sha256_bytes(data)
        try:
            self.client.put_object(
                Bucket=self.bucket,
                Key=key,
                Body=data,
                ContentType=content_type,
                Metadata={"sha256": digest},
            )
            head = self.client.head_object(Bucket=self.bucket, Key=key)
            if int(head.get("ContentLength", -1)) != len(data):
                raise OrchestratorError(f"Remote verification size mismatch after upload: {key}")
            if head.get("Metadata", {}).get("sha256") != digest:
                raise OrchestratorError(f"Remote verification checksum metadata missing/mismatched: {key}")
        except OrchestratorError:
            raise
        except Exception as e:
            raise OrchestratorError(f"DagsHub PUT failed for '{key}': {e}") from e

    def upload_file(self, local_path: Path, key: str, sha256: str) -> Dict[str, Any]:
        size = local_path.stat().st_size
        content_type = mimetypes.guess_type(local_path.name)[0] or "application/octet-stream"
        try:
            self.client.upload_file(
                str(local_path), self.bucket, key,
                ExtraArgs={"ContentType": content_type, "Metadata": {"sha256": sha256}},
            )
            head = self.client.head_object(Bucket=self.bucket, Key=key)
            if int(head.get("ContentLength", -1)) != size:
                raise OrchestratorError(f"Remote size verification failed for {key}")
            if head.get("Metadata", {}).get("sha256") != sha256:
                raise OrchestratorError(
                    f"Remote SHA-256 metadata verification failed for {key}. "
                    "The object is not considered complete."
                )
        except OrchestratorError:
            raise
        except Exception as e:
            raise OrchestratorError(f"DagsHub upload failed for '{key}': {e}") from e
        return {"key": key, "size": size, "sha256": sha256}

    def verify_object(self, record: Dict[str, Any]) -> bool:
        try:
            head = self.client.head_object(Bucket=self.bucket, Key=str(record["key"]))
            return (
                int(head.get("ContentLength", -1)) == int(record["size"])
                and head.get("Metadata", {}).get("sha256") == record["sha256"]
            )
        except Exception as e:
            if self._not_found(e):
                return False
            return False


def make_store(config: Dict[str, Any], execute: bool, workspace: Path) -> ArtifactStore:
    storage = config.get("storage", {})
    kind = storage.get("type", "dagshub_bucket")
    if not execute:
        # Dry-run deliberately avoids remote writes/authentication.
        return LocalArtifactStore(workspace / "dry_run_store")
    if kind == "dagshub_bucket":
        return DagshubBucketStore(str(storage.get("repo", "")))
    if kind == "local":
        raw_root = str(storage.get("local_root", "./jc_artifacts"))
        root = resolve_path(raw_root, workspace)
        return LocalArtifactStore(root)
    raise OrchestratorError(f"Unknown storage.type '{kind}'. Use 'dagshub_bucket' or 'local'.")


def get_remote_json(store: ArtifactStore, key: str) -> Optional[Dict[str, Any]]:
    raw = store.get_bytes(key)
    if raw is None:
        return None
    try:
        value = json.loads(raw.decode("utf-8"))
    except Exception as e:
        raise OrchestratorError(f"Stored JSON is invalid at '{key}': {e}") from e
    if not isinstance(value, dict):
        raise OrchestratorError(f"Stored JSON at '{key}' is not an object.")
    return value


def put_json(store: ArtifactStore, key: str, value: Any) -> None:
    store.put_bytes(key, json.dumps(value, indent=2, ensure_ascii=False).encode("utf-8"), "application/json")


def load_config(path: Path) -> Tuple[Dict[str, Any], Path]:
    cfg = load_json(path)
    if not isinstance(cfg, dict):
        raise OrchestratorError("Config must be a JSON object.")
    base_dir = path.resolve().parent
    cfg.setdefault("project_id", "julius_caesar")
    cfg.setdefault("project_title", "Julius Caesar")
    cfg.setdefault("total_scenes", 18)
    cfg.setdefault("scene_names", {})
    cfg.setdefault("style_prompt", "")
    cfg.setdefault("negative_prompt", "")
    cfg.setdefault("pipeline_version", "v1")
    cfg.setdefault("base_seed", 123456)
    cfg.setdefault("poll_seconds", 3)
    cfg.setdefault("job_timeout_seconds", 7200)
    cfg.setdefault("minimum_output_bytes", 100)
    cfg.setdefault("max_attempts_per_shot", 2)
    cfg.setdefault("cleanup_after_each_shot", False)
    cfg.setdefault("cleanup_payload", {"unload_models": True, "free_memory": True})
    cfg.setdefault("workspace_dir", "./jc_workspace")
    cfg.setdefault("comfyui_output_dir", "/content/ComfyUI/output")
    cfg.setdefault("comfyui_servers", [{"name": "primary", "base_url": "http://127.0.0.1:8188", "output_dir": cfg["comfyui_output_dir"]}])
    cfg.setdefault("workflow_template", "workflow_api.json")
    cfg.setdefault("field_bindings", [])
    cfg.setdefault("storage", {"type": "dagshub_bucket", "repo": "dpbasvs-afk/comfyui-project", "key_prefix": "production_runs"})
    cfg["_base_dir"] = str(base_dir)
    cfg["_config_path"] = str(path.resolve())
    cfg["_workflow_path"] = str(resolve_path(str(cfg["workflow_template"]), base_dir))
    cfg["_workspace_path"] = str(resolve_path(str(cfg["workspace_dir"]), base_dir))
    # Normalize worker specifications.
    servers = cfg.get("comfyui_servers")
    if not isinstance(servers, list) or not servers:
        raise OrchestratorError("comfyui_servers must be a non-empty list.")
    normalized = []
    for i, server in enumerate(servers, start=1):
        if isinstance(server, str):
            server = {"name": f"worker_{i}", "base_url": server, "output_dir": cfg["comfyui_output_dir"]}
        if not isinstance(server, dict) or not server.get("base_url"):
            raise OrchestratorError(f"comfyui_servers item {i} needs base_url.")
        output = str(server.get("output_dir", cfg["comfyui_output_dir"]))
        normalized.append({
            "name": str(server.get("name", f"worker_{i}")),
            "base_url": str(server["base_url"]).rstrip("/"),
            "output_dir": str(resolve_path(output, base_dir)),
        })
    cfg["comfyui_servers"] = normalized
    try:
        cfg["total_scenes"] = int(cfg["total_scenes"])
        if cfg["total_scenes"] < 1:
            raise ValueError
    except (TypeError, ValueError):
        raise OrchestratorError("total_scenes must be a positive integer.")
    if not isinstance(cfg["field_bindings"], list):
        raise OrchestratorError("field_bindings must be a list.")
    return cfg, base_dir


def validate_workflow_config(cfg: Dict[str, Any], require_files: bool = True) -> Tuple[Optional[Dict[str, Any]], List[str]]:
    problems: List[str] = []
    workflow_path = Path(cfg["_workflow_path"])
    workflow: Optional[Dict[str, Any]] = None
    if require_files:
        if not workflow_path.is_file():
            problems.append(f"Workflow template missing: {workflow_path}")
        else:
            try:
                workflow = load_json(workflow_path)
                if not isinstance(workflow, dict) or not workflow:
                    problems.append("Workflow template must be a non-empty ComfyUI API-format JSON object.")
                    workflow = None
            except OrchestratorError as e:
                problems.append(str(e))
    bindings = cfg.get("field_bindings", [])
    if require_files and not bindings:
        problems.append("field_bindings is empty; at least one prompt/shot_prompt input binding is required.")
    for i, binding in enumerate(bindings, start=1):
        if not isinstance(binding, dict):
            problems.append(f"field_bindings[{i}] must be an object")
            continue
        for key in ("node_id", "field", "source"):
            if not binding.get(key):
                problems.append(f"field_bindings[{i}] is missing '{key}'")
        if binding.get("source") and binding.get("source") not in VALID_SOURCES:
            problems.append(f"field_bindings[{i}] has unsupported source '{binding.get('source')}'")
    if require_files and bindings and not any(b.get("source") in {"prompt", "shot_prompt"} for b in bindings if isinstance(b, dict)):
        problems.append("At least one field binding must use source 'prompt' or 'shot_prompt'.")
    if require_files and workflow is not None:
        for i, binding in enumerate(cfg.get("field_bindings", []), start=1):
            node_id = str(binding.get("node_id", ""))
            field = str(binding.get("field", ""))
            node = workflow.get(node_id)
            if not isinstance(node, dict):
                problems.append(f"Binding {i}: node_id '{node_id}' not found in workflow JSON")
            elif not isinstance(node.get("inputs"), dict) or field not in node["inputs"]:
                problems.append(f"Binding {i}: field '{field}' not found in workflow node '{node_id}' inputs")
    return workflow, problems


def scene_id_for(number: int) -> str:
    return f"scene_{number:03d}"


def get_scene_name(cfg: Dict[str, Any], number: int) -> str:
    names = cfg.get("scene_names", {})
    if isinstance(names, list) and number <= len(names):
        return str(names[number - 1] or f"Scene {number}")
    if isinstance(names, dict):
        return str(names.get(str(number), names.get(number, f"Scene {number}")))
    return f"Scene {number}"


def scene_plan_key(cfg: Dict[str, Any], scene_id: str) -> str:
    prefix = str(cfg.get("storage", {}).get("key_prefix", "production_runs")).strip("/")
    project = safe_slug(str(cfg.get("project_id", "julius_caesar")))
    return f"{prefix}/{project}/plans/{scene_id}.json"


def shot_metadata_key(cfg: Dict[str, Any], scene_id: str, shot_id: str) -> str:
    prefix = str(cfg.get("storage", {}).get("key_prefix", "production_runs")).strip("/")
    project = safe_slug(str(cfg.get("project_id", "julius_caesar")))
    return f"{prefix}/{project}/scenes/{scene_id}/shots/{shot_id}/shot.json"


def shot_output_key(cfg: Dict[str, Any], scene_id: str, shot_id: str, index: int, filename: str) -> str:
    prefix = str(cfg.get("storage", {}).get("key_prefix", "production_runs")).strip("/")
    project = safe_slug(str(cfg.get("project_id", "julius_caesar")))
    base = safe_slug(Path(filename).stem, "artifact")
    ext = Path(filename).suffix[:20]
    return f"{prefix}/{project}/scenes/{scene_id}/shots/{shot_id}/outputs/artifact_{index:02d}_{base}{ext}"


def build_prompt(shot: Shot, scene: Dict[str, Any], cfg: Dict[str, Any]) -> str:
    parts = []
    for label, value in (
        ("Project style", str(cfg.get("style_prompt", "")).strip()),
        ("Scene context", str(scene.get("notes", "")).strip()),
        ("Characters", shot.characters),
        ("Shot description", shot.description),
        ("Camera", shot.camera),
        ("Continuity", shot.continuity),
    ):
        if value:
            parts.append(f"{label}: {value}")
    return "\n".join(parts)


def seed_for(cfg: Dict[str, Any], scene_id: str, shot: Shot) -> int:
    base = int(cfg.get("base_seed", 123456))
    digest = hashlib.sha256(f"{scene_id}/{shot.shot_id}".encode("utf-8")).digest()
    offset = int.from_bytes(digest[:8], "big") % 1_000_000_000
    return (base + offset) % (2**32 - 1)


def shot_values(shot: Shot, scene: Dict[str, Any], cfg: Dict[str, Any], shot_index: int) -> Dict[str, Any]:
    sid = str(scene["scene_id"])
    prompt = build_prompt(shot, scene, cfg)
    negative = shot.negative_prompt or str(cfg.get("negative_prompt", ""))
    seed = seed_for(cfg, sid, shot)
    return {
        "prompt": prompt,
        "shot_prompt": shot.description,
        "negative_prompt": negative,
        "dialogue": shot.dialogue,
        "characters": shot.characters,
        "camera": shot.camera,
        "continuity": shot.continuity,
        "scene_id": sid,
        "scene_name": str(scene["name"]),
        "shot_id": shot.shot_id,
        "shot_number": shot_index,
        "shot_index": shot_index,
        "shot_count": int(scene.get("shot_count", 0)),
        "seed": seed,
        "duration_seconds": shot.duration_seconds,
        "filename_prefix": f"JC/{sid}/{shot.shot_id}",
    }


def apply_bindings(template: Dict[str, Any], bindings: List[Dict[str, Any]], values: Dict[str, Any]) -> Dict[str, Any]:
    workflow = copy.deepcopy(template)
    for i, binding in enumerate(bindings, start=1):
        node_id = str(binding["node_id"])
        field = str(binding["field"])
        source = str(binding["source"])
        if source not in values:
            raise OrchestratorError(f"Binding {i}: unsupported source '{source}'.")
        node = workflow.get(node_id)
        if not isinstance(node, dict) or not isinstance(node.get("inputs"), dict) or field not in node["inputs"]:
            raise OrchestratorError(f"Binding {i} no longer matches the workflow: node '{node_id}', field '{field}'.")
        value = values[source]
        if "format" in binding:
            value = str(binding["format"]).format(value=value, **values)
        node["inputs"][field] = value
    return workflow


def input_fingerprint(cfg: Dict[str, Any], workflow_bytes: bytes, scene: Dict[str, Any], shot: Shot, values: Dict[str, Any]) -> str:
    payload = {
        "runner_version": VERSION,
        "pipeline_version": cfg.get("pipeline_version", "v1"),
        "workflow_sha256": sha256_bytes(workflow_bytes),
        "field_bindings": cfg.get("field_bindings", []),
        "style_prompt": cfg.get("style_prompt", ""),
        "negative_prompt": cfg.get("negative_prompt", ""),
        "scene": {k: scene.get(k) for k in ("scene_id", "name", "notes")},
        "shot": asdict(shot),
        "render_values": values,
    }
    return sha256_bytes(canonical_json_bytes(payload))


def is_verified_complete(store: ArtifactStore, cfg: Dict[str, Any], scene_id: str, shot: Shot, fingerprint: str) -> bool:
    record = get_remote_json(store, shot_metadata_key(cfg, scene_id, shot.shot_id))
    if not record or record.get("status") != "completed" or record.get("fingerprint") != fingerprint:
        return False
    outputs = record.get("outputs")
    if not isinstance(outputs, list) or not outputs:
        return False
    return all(isinstance(item, dict) and store.verify_object(item) for item in outputs)


def http_json(url: str, method: str = "GET", payload: Optional[Dict[str, Any]] = None, timeout: float = 15.0) -> Dict[str, Any]:
    body = None if payload is None else json.dumps(payload).encode("utf-8")
    headers = {"Accept": "application/json"}
    if body is not None:
        headers["Content-Type"] = "application/json"
    request = urllib.request.Request(url, data=body, headers=headers, method=method)
    try:
        with urllib.request.urlopen(request, timeout=timeout) as response:
            raw = response.read()
            return json.loads(raw.decode("utf-8")) if raw else {}
    except urllib.error.HTTPError as e:
        detail = e.read().decode("utf-8", errors="replace")[:2000]
        raise OrchestratorError(f"ComfyUI HTTP {e.code} at {url}: {detail}") from e
    except (urllib.error.URLError, TimeoutError, OSError, json.JSONDecodeError) as e:
        raise OrchestratorError(f"ComfyUI request failed at {url}: {e}") from e


def preflight_server(server: Dict[str, str], workflow: Dict[str, Any], cfg: Dict[str, Any]) -> None:
    base = server["base_url"]
    http_json(f"{base}/system_stats", timeout=10)
    node_info = http_json(f"{base}/object_info", timeout=30)
    missing = sorted({str(node.get("class_type")) for node in workflow.values()
                      if isinstance(node, dict) and node.get("class_type") and str(node.get("class_type")) not in node_info})
    if missing:
        raise OrchestratorError(
            f"Worker '{server['name']}' is missing {len(missing)} workflow node type(s): "
            + ", ".join(missing[:20])
            + (" ..." if len(missing) > 20 else "")
            + ". Install/verify extensions before execution."
        )
    output_dir = Path(server["output_dir"])
    if not output_dir.exists():
        raise OrchestratorError(f"Output directory for worker '{server['name']}' does not exist: {output_dir}")


def submit_prompt(server: Dict[str, str], workflow: Dict[str, Any]) -> str:
    base = server["base_url"]
    response = http_json(f"{base}/prompt", method="POST", payload={"prompt": workflow}, timeout=60)
    if response.get("node_errors"):
        raise OrchestratorError(f"ComfyUI rejected workflow on {server['name']}: {json.dumps(response['node_errors'])[:2000]}")
    prompt_id = response.get("prompt_id")
    if not prompt_id:
        raise OrchestratorError(f"ComfyUI did not return prompt_id on {server['name']}: {response}")
    return str(prompt_id)


def wait_for_history(server: Dict[str, str], prompt_id: str, poll_seconds: float, timeout_seconds: float) -> Dict[str, Any]:
    deadline = time.monotonic() + timeout_seconds
    last_note = 0.0
    while time.monotonic() < deadline:
        history = http_json(f"{server['base_url']}/history/{prompt_id}", timeout=30)
        entry = history.get(prompt_id) if isinstance(history, dict) else None
        if isinstance(entry, dict):
            status = entry.get("status", {})
            status_str = str(status.get("status_str", "")).lower() if isinstance(status, dict) else ""
            if status_str in {"error", "failed"}:
                messages = status.get("messages", []) if isinstance(status, dict) else []
                raise OrchestratorError(f"ComfyUI job {prompt_id} failed on {server['name']}: {str(messages)[:3000]}")
            # Standard ComfyUI history is populated after execution; use this entry as completion evidence.
            if isinstance(entry.get("outputs"), dict):
                return entry
        if time.monotonic() - last_note >= 30:
            print(f"[{server['name']}] Waiting for prompt {prompt_id}...")
            last_note = time.monotonic()
        time.sleep(max(0.5, poll_seconds))
    raise OrchestratorError(f"Timed out waiting for prompt {prompt_id} on {server['name']} after {timeout_seconds}s.")


def extract_output_refs(outputs: Any) -> List[Dict[str, str]]:
    refs: List[Dict[str, str]] = []
    def visit(item: Any) -> None:
        if isinstance(item, dict):
            filename = item.get("filename")
            typ = item.get("type", "output")
            subfolder = item.get("subfolder", "")
            if isinstance(filename, str) and filename and typ == "output":
                refs.append({"filename": filename, "subfolder": str(subfolder), "type": str(typ)})
            for value in item.values():
                visit(value)
        elif isinstance(item, list):
            for value in item:
                visit(value)
    visit(outputs)
    # Avoid duplicate references that some nodes expose in more than one output field.
    unique = []
    seen = set()
    for ref in refs:
        marker = (ref["subfolder"], ref["filename"], ref["type"])
        if marker not in seen:
            seen.add(marker)
            unique.append(ref)
    return unique


def output_local_path(server: Dict[str, str], ref: Dict[str, str]) -> Path:
    root = Path(server["output_dir"]).resolve()
    subfolder = PurePosixPath(ref.get("subfolder", ""))
    filename = PurePosixPath(ref["filename"])
    if filename.is_absolute() or ".." in filename.parts or ".." in subfolder.parts or subfolder.is_absolute():
        raise OrchestratorError(f"Unsafe output path returned by ComfyUI: {ref}")
    path = (root / Path(*subfolder.parts) / Path(*filename.parts)).resolve()
    if root not in path.parents:
        raise OrchestratorError(f"ComfyUI output path escaped output directory: {path}")
    return path


def request_memory_cleanup(server: Dict[str, str], cfg: Dict[str, Any]) -> None:
    """Optionally request ComfyUI model/memory cleanup without restarting the server.

    This is disabled by default to avoid repeatedly unloading and reloading models.
    ComfyUI's /free endpoint is a best-effort request, not a guarantee that all GPU
    memory is empty. If unload_models=True, models may need to reload from local disk.
    """
    if not cfg.get("cleanup_after_each_shot", False):
        return
    payload = cfg.get("cleanup_payload", {"unload_models": True, "free_memory": True})
    try:
        request = urllib.request.Request(
            f"{server['base_url']}/free",
            data=json.dumps(payload).encode("utf-8"),
            headers={"Content-Type": "application/json", "Accept": "application/json"},
            method="POST",
        )
        with urllib.request.urlopen(request, timeout=15) as response:
            if response.status not in (200, 202, 204):
                print(f"WARNING: ComfyUI cleanup request returned HTTP {response.status}.")
            else:
                print(f"[{server['name']}] Requested ComfyUI memory cleanup (server remains running).")
    except Exception as e:
        print(f"WARNING: memory cleanup request failed (non-fatal): {e}")


def render_one(server: Dict[str, str], workflow: Dict[str, Any], shot: Shot, scene: Dict[str, Any], cfg: Dict[str, Any], store: ArtifactStore, fingerprint: str) -> Dict[str, Any]:
    print(f"\n[{server['name']}] Submitting {scene['scene_id']}/{shot.shot_id}")
    prompt_id = submit_prompt(server, workflow)
    entry = wait_for_history(server, prompt_id, float(cfg["poll_seconds"]), float(cfg["job_timeout_seconds"]))
    refs = extract_output_refs(entry.get("outputs", {}))
    if not refs:
        raise OrchestratorError(
            f"Job {prompt_id} finished but ComfyUI history exposed no saved output files. "
            "Check that the workflow contains the intended Save Image/Save Video node and that it saves to output."
        )

    outputs: List[Dict[str, Any]] = []
    min_bytes = int(cfg.get("minimum_output_bytes", 100))
    for index, ref in enumerate(refs, start=1):
        local_path = output_local_path(server, ref)
        if not local_path.is_file():
            raise OrchestratorError(f"ComfyUI reported an output that is missing on disk: {local_path}")
        size = local_path.stat().st_size
        if size < min_bytes:
            raise OrchestratorError(f"Output looks incomplete (size {size} bytes): {local_path}")
        digest = sha256_file(local_path)
        key = shot_output_key(cfg, scene["scene_id"], shot.shot_id, index, local_path.name)
        item = store.upload_file(local_path, key, digest)
        item["source_filename"] = local_path.name
        item["comfy_subfolder"] = ref.get("subfolder", "")
        outputs.append(item)
        print(f"[{server['name']}] Verified artifact {index}/{len(refs)}: {key} ({size:,} bytes)")

    record = {
        "schema_version": 1,
        "runner_version": VERSION,
        "status": "completed",
        "project_id": cfg["project_id"],
        "scene_id": scene["scene_id"],
        "scene_name": scene["name"],
        "shot_id": shot.shot_id,
        "prompt_id": prompt_id,
        "fingerprint": fingerprint,
        "pipeline_version": cfg.get("pipeline_version", "v1"),
        "completed_at_utc": now_utc(),
        "worker": server["name"],
        "outputs": outputs,
    }
    metadata_key = shot_metadata_key(cfg, scene["scene_id"], shot.shot_id)
    put_json(store, metadata_key, record)  # written last: acts as the completion commit marker
    stored = get_remote_json(store, metadata_key)
    if not stored or stored.get("fingerprint") != fingerprint or not all(store.verify_object(x) for x in outputs):
        raise OrchestratorError(f"Final completion marker or remote artifact verification failed for {shot.shot_id}.")
    print(f"[{server['name']}] COMPLETE: {scene['scene_id']}/{shot.shot_id}")
    return record


def build_plan(cfg: Dict[str, Any], scene_num: int, scene_name: str, scene_notes: str, shots: List[Shot]) -> Dict[str, Any]:
    return {
        "schema_version": 1,
        "project_id": cfg["project_id"],
        "project_title": cfg["project_title"],
        "total_scenes": cfg["total_scenes"],
        "scene": {
            "scene_number": scene_num,
            "scene_id": scene_id_for(scene_num),
            "name": scene_name,
            "notes": scene_notes,
            "shot_count": len(shots),
        },
        "shots": [asdict(s) for s in shots],
        "created_at_utc": now_utc(),
        "runner_version": VERSION,
    }


def choose_plan(cfg: Dict[str, Any], store: ArtifactStore, scene_num: int, shot_file: Optional[Path], workspace: Path) -> Dict[str, Any]:
    sid = scene_id_for(scene_num)
    local_plan_path = workspace / "plans" / f"{sid}.json"
    key = scene_plan_key(cfg, sid)
    existing: Optional[Dict[str, Any]] = None
    if shot_file is None:
        # Execution mode can resume a plan from the remote bucket; dry-run consults only local plan files.
        if local_plan_path.is_file():
            existing = load_json(local_plan_path)
        elif isinstance(store, DagshubBucketStore):
            existing = get_remote_json(store, key)
        if existing:
            existing_scene = existing.get("scene", {})
            count = len(existing.get("shots", []))
            print(f"Found an existing plan for {sid}: {count} shot(s), scene name '{existing_scene.get('name', sid)}'.")
            choice = ask("Reuse this shot plan? Y=yes, N=enter a new plan", "Y").lower()
            if choice in {"y", "yes", ""}:
                # Keep a local cache for convenience and ensure plans created during
                # dry-run are promoted to DagsHub when the user later chooses execution.
                local_plan_path.parent.mkdir(parents=True, exist_ok=True)
                local_plan_path.write_text(json.dumps(existing, indent=2, ensure_ascii=False), encoding="utf-8")
                if isinstance(store, DagshubBucketStore):
                    put_json(store, key, existing)
                return existing
            if choice not in {"n", "no"}:
                raise OrchestratorError("Choose Y or N.")

    if shot_file is not None:
        shots = read_shot_file(shot_file)
        default_name = get_scene_name(cfg, scene_num)
        scene_name = default_name
        scene_notes = ""
    else:
        scene_name = ask("Scene name", get_scene_name(cfg, scene_num))
        scene_notes = ask("Optional scene-level continuity/context (press Enter to skip)", "")
        shot_count = ask_int("Number of shots in this scene", minimum=1)
        shots = gather_shots_interactive(shot_count)
    ensure_unique_shot_ids(shots)
    plan = build_plan(cfg, scene_num, scene_name, scene_notes, shots)
    local_plan_path.parent.mkdir(parents=True, exist_ok=True)
    local_plan_path.write_text(json.dumps(plan, indent=2, ensure_ascii=False), encoding="utf-8")
    if isinstance(store, DagshubBucketStore):
        put_json(store, key, plan)
        print(f"Saved scene plan to DagsHub Storage: {key}")
    else:
        print(f"Saved local scene plan: {local_plan_path}")
    return plan


def print_plan(plan: Dict[str, Any], config: Dict[str, Any], execute: bool) -> None:
    scene = plan["scene"]
    print("\n" + "=" * 68)
    print("JULIUS CAESAR — SCENE/SHOT ORCHESTRATOR")
    print("=" * 68)
    print(f"Project      : {config['project_title']} ({config['project_id']})")
    print(f"Scene        : {scene['scene_id']} — {scene['name']}")
    print(f"Shots        : {len(plan['shots'])}")
    print(f"ComfyUI URLs : {', '.join(s['base_url'] for s in config['comfyui_servers'])}")
    print(f"Mode         : {'LIVE EXECUTION' if execute else 'DRY RUN — no jobs submitted'}")
    print("-" * 68)
    for i, raw in enumerate(plan["shots"], start=1):
        shot = Shot.from_obj(raw, i)
        print(f"{i:03d}. {shot.shot_id}: {shot.description.replace(chr(10), ' ')[:180]}")
    print("=" * 68)


def run_scene(cfg: Dict[str, Any], plan: Dict[str, Any], execute: bool, store: ArtifactStore, workflow: Optional[Dict[str, Any]]) -> int:
    scene = dict(plan["scene"])
    sid = str(scene["scene_id"])
    shots = [Shot.from_obj(raw, i) for i, raw in enumerate(plan["shots"], start=1)]
    scene["shot_count"] = len(shots)
    workspace = Path(cfg["_workspace_path"])
    workspace.mkdir(parents=True, exist_ok=True)

    if not execute:
        workflow_bytes = Path(cfg["_workflow_path"]).read_bytes() if Path(cfg["_workflow_path"]).is_file() else b"workflow-template-not-yet-provided"
        print("\nDRY RUN: no ComfyUI request and no Dagshub write will be made.")
        for i, shot in enumerate(shots, start=1):
            values = shot_values(shot, scene, cfg, i)
            fingerprint = input_fingerprint(cfg, workflow_bytes, scene, shot, values)
            print(f"WOULD PROCESS {sid}/{shot.shot_id} seed={values['seed']} fingerprint={fingerprint[:12]} prompt={values['prompt'][:140]!r}")
        print("\nTo execute later: add --execute after supplying a valid API workflow and node bindings.")
        return 0

    if workflow is None:
        raise OrchestratorError("Internal error: live execution has no loaded workflow.")
    workflow_bytes = Path(cfg["_workflow_path"]).read_bytes()
    # Preflight every worker before sending any jobs.
    for server in cfg["comfyui_servers"]:
        print(f"Preflight worker '{server['name']}' at {server['base_url']}...")
        preflight_server(server, workflow, cfg)
        # Also verify bindings at runtime against exact API graph inputs.
        for binding in cfg["field_bindings"]:
            node_id, field = str(binding["node_id"]), str(binding["field"])
            node = workflow.get(node_id)
            if not isinstance(node, dict) or field not in node.get("inputs", {}):
                raise OrchestratorError(f"Invalid workflow binding: node {node_id}, input {field}.")
        print(f"Worker '{server['name']}' ready.")

    pending: List[Tuple[Shot, Dict[str, Any], str]] = []
    skipped = 0
    for i, shot in enumerate(shots, start=1):
        values = shot_values(shot, scene, cfg, i)
        fingerprint = input_fingerprint(cfg, workflow_bytes, scene, shot, values)
        if is_verified_complete(store, cfg, sid, shot, fingerprint):
            print(f"SKIP verified existing output: {sid}/{shot.shot_id}")
            skipped += 1
        else:
            pending.append((shot, values, fingerprint))

    if not pending:
        print(f"\nScene already complete: {skipped}/{len(shots)} shot(s) verified. Nothing to render.")
        return 0

    servers = cfg["comfyui_servers"]
    queues: List[List[Tuple[Shot, Dict[str, Any], str]]] = [[] for _ in servers]
    for index, item in enumerate(pending):
        queues[index % len(servers)].append(item)
    errors: List[str] = []
    completed = 0

    def worker_loop(server: Dict[str, str], queue: List[Tuple[Shot, Dict[str, Any], str]]) -> Tuple[int, List[str]]:
        local_completed = 0
        local_errors: List[str] = []
        max_attempts = max(1, int(cfg.get("max_attempts_per_shot", 2)))
        for shot, values, fingerprint in queue:
            success = False
            last_error: Optional[Exception] = None
            rendered_workflow = apply_bindings(workflow, cfg["field_bindings"], values)
            for attempt in range(1, max_attempts + 1):
                try:
                    if attempt > 1:
                        print(f"[{server['name']}] Retry {attempt}/{max_attempts} for {sid}/{shot.shot_id}")
                        time.sleep(min(5 * attempt, 20))
                    render_one(server, rendered_workflow, shot, scene, cfg, store, fingerprint)
                    request_memory_cleanup(server, cfg)
                    local_completed += 1
                    success = True
                    break
                except Exception as e:
                    last_error = e
                    print(f"WARNING: attempt {attempt}/{max_attempts} failed for {sid}/{shot.shot_id}: {e}")
            if not success:
                message = f"{sid}/{shot.shot_id} on {server['name']}: {last_error}"
                local_errors.append(message)
                print(f"ERROR: {message}")
                # Continue processing other shots. Rerunning later will retry this shot.
        return local_completed, local_errors

    with ThreadPoolExecutor(max_workers=len(servers)) as pool:
        futures = [pool.submit(worker_loop, server, queue) for server, queue in zip(servers, queues) if queue]
        for future in as_completed(futures):
            done, worker_errors = future.result()
            completed += done
            errors.extend(worker_errors)

    print("\n" + "=" * 68)
    print("SCENE RUN SUMMARY")
    print(f"Scene: {sid} — {scene['name']}")
    print(f"Already verified/skipped : {skipped}")
    print(f"Newly completed          : {completed}")
    print(f"Failed this run           : {len(errors)}")
    print(f"Total shots in plan       : {len(shots)}")
    if errors:
        print("\nFailures (rerun the same scene to retry only shots that are not verified):")
        for error in errors:
            print(" - " + error)
    print("=" * 68)
    return 1 if errors else 0


def main(argv: Optional[Sequence[str]] = None) -> int:
    parser = argparse.ArgumentParser(description="ComfyUI scene-and-shot orchestrator; dry-run unless --execute is supplied.")
    parser.add_argument("--config", default="jc_orchestrator_config.json", help="Path to runner config JSON")
    parser.add_argument("--scene", type=int, help="Scene number to process; interactive selection if omitted")
    parser.add_argument("--shots-file", help="JSON or TXT shot breakdown file for the selected scene")
    parser.add_argument("--execute", action="store_true", help="Actually submit jobs and write outputs to configured storage")
    parser.add_argument("--validate-only", action="store_true", help="Validate local config and workflow structure; no jobs or remote writes")
    args = parser.parse_args(argv)

    try:
        config_path = Path(args.config).expanduser().resolve()
        cfg, _ = load_config(config_path)
        workspace = Path(cfg["_workspace_path"])

        if args.validate_only:
            workflow, problems = validate_workflow_config(cfg, require_files=True)
            print(f"Config: {config_path}")
            print(f"Workflow: {cfg['_workflow_path']}")
            print(f"Servers configured: {len(cfg['comfyui_servers'])}")
            if problems:
                print("VALIDATION FAILED:")
                for item in problems:
                    print(" -", item)
                print("Note: node IDs/fields cannot be assumed; they must match the exported API workflow exactly.")
                return 2
            print(f"VALIDATION OK: {len(workflow)} workflow nodes; {len(cfg['field_bindings'])} input bindings.")
            print("This validates local files only; it does not prove extension or live-server compatibility.")
            return 0

        if args.scene is not None:
            scene_num = args.scene
            if scene_num < 1 or scene_num > cfg["total_scenes"]:
                raise OrchestratorError(f"Scene must be between 1 and {cfg['total_scenes']}.")
        else:
            print(f"Project '{cfg['project_title']}' is configured for {cfg['total_scenes']} scene(s).")
            scene_num = ask_int("Which scene number should be processed", 1, cfg["total_scenes"])

        shots_file = Path(args.shots_file).expanduser().resolve() if args.shots_file else None
        workflow = None
        if args.execute:
            workflow, problems = validate_workflow_config(cfg, require_files=True)
            if problems:
                raise OrchestratorError(
                    "Cannot execute yet because the API workflow/config is incomplete:\n - " + "\n - ".join(problems)
                )
        store = make_store(cfg, execute=args.execute, workspace=workspace)
        try:
            plan = choose_plan(cfg, store, scene_num, shots_file, workspace)
            print_plan(plan, cfg, args.execute)
            if not args.execute:
                return run_scene(cfg, plan, execute=False, store=store, workflow=None)
            return run_scene(cfg, plan, execute=True, store=store, workflow=workflow)
        finally:
            store.close()
    except (OrchestratorError, KeyboardInterrupt) as e:
        if isinstance(e, KeyboardInterrupt):
            print("\nInterrupted by user. Completed shot records remain in storage; rerun to resume.", file=sys.stderr)
            return 130
        print(f"\nERROR: {e}", file=sys.stderr)
        return 2


if __name__ == "__main__":
    raise SystemExit(main())
```
